3. What is a kernel, and why is it important?

Definition: A kernel is the core part of an operating system that acts as a bridge between software and computer hardware.

Why is the kernel important?
-  Process Management: Manages running processes.
2. Memory Management: Allocates and manages computer memory.
3. Device Management: Communicates with hardware devices.
4. System Calls: Allows applications to request services from the operating system.
5. Resource Management: Allocates CPU time and other system resources.

 4.What are the main responsibilities of the Linux kernel?

Main responsibilities
1. Process Management: Creates, schedules, and manages running processes.
2. Memory Management: Allocates and frees RAM for programs.
3. Device Management: Controls hardware devices through device drivers.
4. File System Management: Supports access to files and file systems.
5. CPU Scheduling: Decides which process gets CPU time.
6. Security and Access Control: Helps enforce permissions and isolate processes.
7. Networking: Manages network communication and protocols.
8. System Calls: Provides an interface for applications to request operating system services.

8. What is the difference between a command, an option, and an argument in Linux?


- Definition: In Linux, a command tells the system what to do, an option modifies how the command works, and an argument specifies the target or value on which the command operates.
- Difference between Command, Option, and Argument

- Command=>Tells the system what action to perform.	=>ls
- Option=>Changes or controls the command's behavior.=>	-l
- Argument=>Specifies the file, folder, or value to work with.=>	Documents

- ls -l Documents

| Feature  | Command                         | Option                              | Argument                              |
|----------|---------------------------------|-------------------------------------|---------------------------------------|
| Meaning  | Tells the system what to do.    | Modifies command behavior.          | Specifies the target or value.       |
| Purpose  | Performs an action.             | Controls how the action is performed.| Specifies what to work on.           |
| Syntax   | command                         | command -option                     | command argument                      |
| Example  | ls                              | ls -l                               | ls Documents                          |
| Use      | Lists files and directories.    | Displays detailed file information.| Lists contents of Documents folder.   |

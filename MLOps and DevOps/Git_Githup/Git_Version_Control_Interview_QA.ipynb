# Git and Version Control – Interview Questions & Answers

Beginner-friendly interview preparation notes. Read each question and practise explaining the answer in your own words.

## 1. What is Git? Why is it called SCM or VCS? Who created it?

**Answer:**

- **Git** is a free, open-source tool used to track changes in files, especially source code.
- **SCM** means *Source Control Management*. It helps manage and track changes to source code.
- **VCS** means *Version Control System*. It keeps a history of changes so developers can compare versions or return to an earlier version.
- Git was created by **Linus Torvalds** in 2005.

**Example:** If a change breaks your program, Git helps you find an earlier version of the code.

## 2. What are the main features of Git? Explain what “distributed” means.

**Answer:**

Main features of Git:
1. **Version history:** Records committed changes.
2. **Rollback:** Helps you return to an earlier version.
3. **Branching:** Lets developers work on separate features.
4. **Merging:** Combines changes from different branches.
5. **Collaboration:** Developers can share changes with a remote repository.
6. **Speed:** Most common operations run locally.
7. **Integrity:** Git identifies content using hashes to help detect changes.

**What does distributed mean?**

Git is a **Distributed Version Control System (DVCS)**. Each developer can have a complete copy of the repository and its history on their own computer. Developers can commit changes locally and share them later.

## 3. What problems do developers face when managing code versions manually?

**Answer:**

Without Git, developers may face these problems:
1. Many folders such as `project_final`, `project_final2`, and `project_latest`.
2. Difficulty knowing exactly what changed.
3. Accidental deletion or overwriting of working code.
4. Difficulty restoring an older working version.
5. Conflicts when multiple developers edit the same files.
6. No reliable, shared history of who changed what and when.

**Example:** A folder named `final_final_latest` does not clearly show which version is correct.

## 4. How does Git roll back to an older version? Which command helps you move to a previous commit?

**Answer:**

Git stores a history of commits. You can inspect that history and use it to find an earlier version.

Useful commands:
```bash
git log --oneline
git checkout <commit-id>
```

- `git log --oneline` displays a short list of commits.
- `git checkout <commit-id>` moves your working directory to the selected commit, usually leaving you in a **detached HEAD** state.

**Important:** To undo changes safely, the correct command depends on what you want to undo. `git revert <commit-id>` creates a new commit that reverses an earlier commit. `git reset` can move a branch pointer and needs care.

## 5. Compare manual versioning and Git.

**Answer:**

| Point | Manual versioning | Git |
|---|---|---|
| Version tracking | Uses manually named folders or copies | Records committed changes in history |
| Rollback | Find and copy an older folder | Use commit history and suitable Git commands |
| Team collaboration | Difficult to coordinate changes | Branches and remote repositories support collaboration |
| Storage | Often duplicates many complete folders | Stores repository objects efficiently and tracks snapshots |
| Change details | Hard to see who changed what | Commit history records author and commit details |

## 6. What are the three types of version control systems?

**Answer:**

1. **Local Version Control System (Local VCS):** Keeps version history on one computer. It can help restore earlier versions, but collaboration and computer-failure recovery are limited.
2. **Centralized Version Control System (Centralized VCS):** Stores the main version history on a central server. Developers connect to that server to share changes.
3. **Distributed Version Control System (Distributed VCS):** Each developer has a local copy of the repository and its history. They can work locally and exchange changes with other repositories.

**Example:** Git is a Distributed Version Control System.

## 7. What are the drawbacks of Local VCS and Centralized VCS? How does DVCS overcome them?

**Answer:**

**Local VCS drawbacks**
- History is usually kept on one computer.
- If that computer is lost or damaged, the history may be lost.
- Sharing and teamwork are difficult.

**Centralized VCS drawbacks**
- Developers depend on the central server for many operations.
- If the server is unavailable, some teamwork operations may stop.
- If the central repository is lost and there is no backup, the project history may be lost.

**How DVCS helps**
- Each developer has a local repository with history.
- Many operations, such as commits and viewing history, work offline.
- Copies of the repository can provide additional recovery options.
- Developers can exchange changes when a connection is available.

**Note:** Backups are still important, even when using a DVCS.

## 8. What is the difference between a local and remote repository? How do Git and GitHub work together in a team?

**Answer:**

- **Local repository:** The repository on your own computer. You can make commits and view history locally.
- **Remote repository:** A repository hosted elsewhere so developers can share changes. GitHub can host remote Git repositories.

**Team workflow:**
1. A developer clones the shared GitHub repository.
2. The developer creates a branch and edits files.
3. The developer uses `git add` and `git commit` to save changes locally.
4. The developer uses `git push` to send commits to GitHub.
5. Teammates use `git pull` to fetch and integrate shared changes.
6. The team may open a pull request to review changes before merging them.

**Remember:** Git is the version-control tool; GitHub is an online platform that hosts Git repositories and supports team collaboration.

## Quick Revision

- Git = version-control tool.
- Git was created by Linus Torvalds in 2005.
- VCS types = Local, Centralized, Distributed.
- `git log --oneline` = view short commit history.
- `git revert <commit-id>` = make a new commit that reverses an earlier commit.
- Local repository = on your computer; remote repository = hosted elsewhere.
- GitHub hosts Git repositories and adds collaboration features.

## 1. What is the difference between Git and GitHub?

**Git** is a version control tool installed on your computer. It tracks changes in files and helps you return to earlier versions.

**GitHub** is an online platform that hosts Git repositories and helps people share and collaborate on projects.

**Example:** Git records your code changes locally; GitHub lets you upload and share that code online.

## 2. What is a repository? What is the difference between a public and a private repository?

A **repository (repo)** is a folder where project files and their change history are stored.

| Public repository | Private repository |
|---|---|
| Anyone can view it. | Only the owner and people granted access can view it. |
| Useful for portfolios and open-source projects. | Useful for personal, confidential, or team projects. |

**Note:** Public does not mean everyone can automatically change your repository.

## 3. How do you check whether Git is installed on your system and which version it is?

Open Terminal, Command Prompt, or PowerShell and run:

```bash
git --version
```

If Git is installed, the command displays a version, for example `git version 2.x.x`. If the command is not recognized, install Git or check that it is added to your system PATH.

## 4. Why do we set user.name and user.email in Git? What does the --global option mean?

Git stores a name and email with commits so the author of each change can be identified.

```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```

`--global` applies these settings to your user account for all repositories by default. Without `--global`, running the command inside a repository sets the value only for that repository.

## 5. What does the git init command do?

`git init` initializes a new Git repository in the current folder. It creates the hidden `.git` directory that Git uses to store history and configuration.

```bash
mkdir my-project
cd my-project
git init
```

After this, you can track the folder's files with Git.

## 6. What is the difference between git add . and git commit -m "message"?

- `git add .` stages new and modified files in the current directory and its subdirectories for the next commit.
- `git commit -m "message"` saves the staged changes as a new commit in the local repository.

**Typical order:**

```bash
git add .
git commit -m "Add project files"
```

Adding stages changes; committing records them in the repository history.

## 7. What is the use of git branch -M main and git branch?

- `git branch -M main` renames the current branch to `main`. The `-M` option forces the rename if needed.
- `git branch` lists local branches. The current branch is marked with `*`.

```bash
git branch -M main
git branch
```

## 8. What does git remote add origin <repo url> do? What is "origin"?

`git remote add origin <repo-url>` connects your local repository to a remote repository URL.

```bash
git remote add origin https://github.com/USERNAME/REPOSITORY.git
```

`origin` is the conventional nickname for that remote. It is just a name; you could choose another name. The command connects the repositories but does not upload files by itself.

## 9. What is the difference between git push -u origin main and git push origin main?

Both commands push the local `main` branch to the remote named `origin`.

- `git push origin main` pushes the branch for this command.
- `git push -u origin main` also sets the upstream tracking relationship. Later, you can often use `git push` and `git pull` without specifying the remote and branch.

`-u` means `--set-upstream`.

## 10. What is the difference between git clone and git pull?

- `git clone <url>` downloads a remote repository and its history to create a local copy. It is commonly used when getting a project for the first time.
- `git pull` updates an existing local repository by fetching remote changes and integrating them into the current branch.

```bash
git clone https://github.com/USERNAME/REPOSITORY.git
cd REPOSITORY
git pull
```

**Remember:** Clone = get a local copy; pull = update a copy you already have.

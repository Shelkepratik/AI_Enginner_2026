Write a function my_sum(iterable) that calculates the sum of all elements WITHOUT using a for loop or sum(). Use iter(), next(), a while loop and StopIteration. Test it with a list, a tuple and a range.

```python
def my_sum(iterable):
    it=iter(iterable)
    total=0

    while True:
        try:
            value=next(it)
            total+=value
        except StopIteration:
            break
    return total

# Test with a list
print(my_sum([10, 20, 30]))

# Test with a tuple
print(my_sum((5, 15, 25)))

# Test with a range
print(my_sum(range(1, 6)))


```

Write a generator function weekdays() that yields the names of the days from Monday to Friday, one by one. Print the first two days using next() and the remaining days using a for loop.

```python
def weekdays():
    yield "monday"
    yield "tuesday"
    yield "wednesday"
    yield "thursday"
    yield "friday"


w=weekdays()
print(next(w))
print(next(w))
print("==================")
for i in w:
    print(i)
```

Write a generator function even_numbers(limit) that yields all even numbers from 2 up to limit. Create a generator object for limit = 20 and print all the values using a for loop. Then call next() on it once more and write a comment about the error.

```python
def even_number(limit):
    for i in range(2,limit+1):
        if i % 2==0:
            yield i

e=even_number(20)
for i in e:
    print(i)

print(next(e))
```

Write a generator function table(num) that yields the multiplication table of a number in the format "5 x 1 = 5" up to 10. Print the table of 7 using a for loop.

```python
def table(num):
    for i in range(1,11):
        yield  f"{num}*{i}={num*i}"

for i in table(7):
    print(i)
```


10. Create a list of squares of numbers from 1 to 100000 using list comprehension, and a generator expression for the same squares. Print the memory size of both using sys.getsizeof() and write a comment explaining the difference.

```python
import sys

# List comprehension
squares_list = [x ** 2 for x in range(1, 100001)]

# Generator expression
squares_gen = (x ** 2 for x in range(1, 100001))

print("List memory size:", sys.getsizeof(squares_list), "bytes")
print("Generator memory size:", sys.getsizeof(squares_gen), "bytes")

# Explanation:
# The list stores all 100000 square values in memory.
# The generator produces values one at a time, so it does not
# store all the square values at once. Therefore, it generally
# uses much less memory than the list.
```

## Generators

- python  generators are a simple way of creating iterators
- python generator is nothing but a function.in generator return keyword is not available instead it uses of yield keyword/statement

```python
def gen_demo():
    yield "first statment"
    yield "second statement"
    yield "thrid statement"
```

```python
gen=gen_demo()
```

```python
print(gen)
```

```python
dir(gen)
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
var3 = range(1, 5)
it = iter(var3)

def display():
    yield it
    yield it
    yield it

dis = display()

print(next(dis))
print(next(dis))
```

```python
var3 = range(1, 5)
it = iter(var3)

def display():
    yield next(it)
    yield next(it)
    yield next(it)

dis = display()

print(next(dis))
print(next(dis))
print(next(dis))
```

#### display number 50 to 60

```python
def my_range(start,end):
    for i in range(start,end):
        yield i

gen_obj1=my_range(50,61)
```

```python
for i in gen_obj1:
    print(i)
```

```python
def own_for_loop(iterable):
    itertor=iter(iterable)
    while True:
        try:
            print(next(itertor))
        except StopIteration:
            break

a=[1,2,3]
own_for_loop(a)
```

```python
b=range(1,6,1)
own_for_loop(b)
```

```python
def square(num):
    for i in range(1,num+1):
        yield i**2

s=square(10)
for i in s:
    print(i)
```

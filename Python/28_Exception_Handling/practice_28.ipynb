Write a function split_bill(amount, people) that returns the amount each person has to pay. Handle ZeroDivisionError inside the function. Call it once with (1200, 4) and once with (1200, 0).

```python
def split_bill(amount,people):
    try:
        return amount/people
    except ZeroDivisionError:
        print("Number of people cannot be zero")
print(split_bill(1200,4))
print(split_bill(1200,0))
```

Write a function safe_divide(a, b) that handles both ZeroDivisionError and TypeError using two separate except blocks. Call it with (50, 5), (50, 0) and (50, "5"), and print the result or error message for each call.

```python
def safe_divide(a,b):
    try: 
        result=a/b
        return result
    except ZeroDivisionError as err:
        print(err)
    except TypeError as err1:
        print(err1)
    
```

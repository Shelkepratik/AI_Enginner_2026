## Operator Precedence

```python
result=(2+3)*4
print(result)
```

```python
result1=2+3*4
print(result1)
```

```python
result2=2+3**2 # 3^2=3*3=9 , 2+9=11
print(result2)
```

```python
result3=10-2+3 # 10-2=8 8+3=11
print(result3)
```

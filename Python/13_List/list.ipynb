## List-Container data type

*list-list is one of the container data type in python.*we can stored hetrogenous element(all the element different data type).
*homogenous elements means all the element are of same data type
*list syntax:
list_name=[element1,element2,.....]

*list is represented using square bracket and elements are seperated by comma

#### Hetrogenous element

```python
list1=[10,10.5,"pratik",True]#Hetrogenouns data type
print(list1)
print(type(list1))
```

#### Homogenous element

```python
list2=[10,20,30,40,50]
print(list2)
```

#### list can contain duplicate element

```python
list3=[10,20,30,10,20]
print(list3)
```

#### Repetition of element using * operator

```python
list4=[10,20,30]
print(list4*3)
```

#### To access all the elements of list

```python
list5=[10,20,30,40,50]
print(list5)
```

#### *Indexing

*if we want to access individual element of the list we use indexing
*These are two types:-
1-positive indexing 
2-Negative indexing

#### positive indexing

```python
list6=[10,20,30,40,50]
print(list6)
print(len(list6))
```

```python
print(list6[0])
print(list6[1])
print(list6[2])
print(list6[3])
print(list6[4])
```

#### Negative indexing

```python
print(list6[-1])
print(list6[-2])
print(list6[-3])
print(list6[-4])
print(list6[-5])
```

#### *slicing-To access the subelement/sublist

syntax=list_name=[start,stop,step]

```python
print(list6)
print(list6[0:3])
print(list6[-3:-1])
print(list6[1:4:1])
print(list6[2:])
print(list6[0::2])
print(list6[::1])#print the all element in the list
```

#### iterate  list using for loop

```python
list=[10,20,30,40,50]
for var in list:
    print(var,end=" ")
```

#### iterate list using while loop

```python
list=[10,20,30,40,50]
index=0
while index<len(list):
    print(list[index],end=" ")
    index+=1
```

#### *enumerate function 

enumerate is one of the built in function in python.we can print element along with its index

syntax:for index,var in enumerate(list_name)

```python
list=[10,20,30,40,50]
for index,var in enumerate(list):
    print(index,var)

```

start the index for 1 and any number

```python
list=[10,20,30,40,50]
for index,var in enumerate(list,start=1):
    print(index,var)

```

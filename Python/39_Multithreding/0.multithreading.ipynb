## Multi-threading in python

Types of computer multitasking its there are two multitasking:
- process based multitasking(Application level multitasking)
- Thread based multitasking(program level multitasking)

what is thread?
- thread is used to execute the instruction in the program
- it is the smallest execution unit in the program
- Bydefault we have only one thread in our program.it is called as main thread
- thread will be only one default main thread in the program until we explictly implement multi-threading

```python
import threading

print("hello world")

current_thread=threading.current_thread()

print(current_thread.name) #thread name

print(current_thread.ident)    #thread id 

```

#### without multithreading

```python
from threading import Thread
import time

def addition():
    time.sleep(10)
    num1=10
    num2=20
    result=num1+num2
    print(result)
   

def multiplication():
    time.sleep(10)
    num1=10
    num2=20
    result=num1*num2
    print(result)
    

def division():
    time.sleep(10)
    num1=10
    num2=20
    result=num2/num1
    print(result)
    
start=time.time()
print(start)
addition()
multiplication()
division()
end=time.time()
print(end)

print("total time:",end-start)
```

### with multithreading

```python
from threading import Thread
import time

def addition():
    time.sleep(10)
    num1=10
    num2=20
    result=num1+num2
    print(result)
   

def multiplication():
    time.sleep(10)
    num1=10
    num2=20
    result=num1*num2
    print(result)
    

def division():
    time.sleep(10)
    num1=10
    num2=20
    result=num2/num1
    print(result)
    


```

```python
t1=Thread(target=addition,name="mythread1")
t2=Thread(target=multiplication,name="mythread2")
t3=Thread(target=division,name="mythread3")
```

```python
t1.start()
t2.start()
t3.start()
```

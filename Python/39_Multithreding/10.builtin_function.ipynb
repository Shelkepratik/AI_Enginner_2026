### printing thread name:
- t1=Thread(target=display,name="mythread")
- print(t1.name)#mythread
- print(t1.getName())#mythread

```python
from threading import Thread
def display():
    print("executing display1 by mythread1")

t1=Thread(target=display,name="myThread")
t1.start()
print("The thread name is:",t1.name)
print("another way to print name:",t1.getName())
```

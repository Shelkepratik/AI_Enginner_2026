### *set Comprehension

set comprehension-set comprehension create a new set by iterating over a sequence and applying an expression to each element like list comprhension but we use curly breaces{} and no duplicate in the result.

{expr for item in sequence/iterable if condition}

create a set having squares of numbers from 1 to 5

```python
set1={i**2 for i in range(1,6)}
print(set1)
```

*without comprehension

```python
new_set=set()
for i in range(1,6):
    if i<=3:
        new_set.add(i**2)
print(new_set)
```

*remove the duplicate letter in string

```python
text="aeroaerthoamruseptryua"
new_set1={i.upper() for i in text if i!="o"}
print(new_set1)
```

```python
new_set2=set(text)
print(new_set2)
```

*convert word into uppercase

```python
animals=["cat","dog","fish"]
new_set3={i.upper() for i in animals if len(i)>3}
print(new_set3)
```

```python
new_set4={i.upper() for i in animals if len(i)<=4}
print(new_set4)
```

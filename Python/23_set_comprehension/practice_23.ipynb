1. Using set comprehension, create a set having squares of numbers from 1 to 5 and print it.


```python
new_set={i**2 for i in range(1,6,1)}
print(new_set)
```

Using set comprehension, create a set having squares of only those numbers from 1 to 10 that are less than or equal to 5.

```python
new_set={i**2 for i in range(1,11) if i<=5}
print(new_set)
```

Solve Q2 without comprehension, using a for loop and the add() method.

```python
new_set=set()
for i in range(1,11):
    if i<=5:
        new_set.add(i**2)
print(new_set)
```

Create a list numbers = [1, 2, 2, 3, 3, 3, 4, 4, 5]. Using set comprehension, create a set of unique numbers multiplied by 1

```python
list1=[1,2,2,3,3,3,4,4,5]
new_set={i*1 for i in list1}
print(new_set)
```

Create a string text = "programming". Using set comprehension, create a set of unique letters in uppercase, excluding the letter "g".

```python
text="programming"
new_set={i.upper() for i in text if i!="g"}
print(new_set)
```

Create a list animals = ["cat", "dog", "fish", "tiger"]. Using set comprehension, create a set of animal names in uppercase whose length is more than 3.

```python
animals=['cat','dog','fish','tiger']
new_set={i.upper() for i in animals if len(i)>3}
print(new_set)
```

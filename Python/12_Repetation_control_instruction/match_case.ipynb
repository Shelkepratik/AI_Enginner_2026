In Python, match-case is used to execute different blocks of code based on the value of a variable. It was introduced in Python 3.10.

Match-case with a calculator

```python
a = 10
b = 5
operator = "+"

match operator:
    case "+":
        print(a + b)
    case "-":
        print(a - b)
    case "*":
        print(a * b)
    case "/":
        print(a / b)
    case _:
        print("Invalid operator")
```

```python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case _:
        print("Invalid day")
```

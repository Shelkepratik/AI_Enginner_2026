## Repeation control statement:-
#### Repetition control instructions, also called looping statements, are used to execute a block of code repeatedly as long as a condition is satisfied or for a specified number of times.

#### *type of Repeation control statement:1-while  2-for

*Python provides mainly two repetition control statements:

*for loop – Used when we want to repeat a block for each item in a sequence or for a known number of times.
*while loop – Used when we want to repeat a block as long as a condition is True.


1-while loop:-until condition is true statement inside while block executes repeatedly
     2-else block is optional
     3-if while block terminated abruptly using break statement else block will not execute.

```python
count=0
while count<=5:
    print(count)
    count=count+1
else:
    print("else block is execute")
print("program execute successfully")
```

```python
count=0
while count<=5:
    print("Hello world")
    count=count+1
else:
    print("program terminate")
```

while block condition is false that time else block executed

```python
count=10
while count<=5:
    print(count)
    count=count+1
else:
    print("else block executed")
```

#### *for loop
1-for loop is used to iterate over sequence. 

iteration in string using for loop

```python
name="ganesh"
for var in name:
    print(var.upper())
else:
    print("program ended")
```


 iteration list using for loop

```python
list1=[10,20,30,40,50]
for var in list1:
    print(var)
```

 iteration  tuple using for loop

```python
tuple=(1,2,3,4,5)
for i in tuple:
    print(i)
```

iteration dict using for loop

```python
student={
    "name":"pratik",
    "age":20
}
for i in student.items():
    print(i)
```

*break:we can used break statement in while loop and for loop
*whwn it execute break statement terminate loop immediately and execute the next statment in the program

```python
for i in range(1,11,1):
    if i==5:
        break
    print(i)
```

```python
count=1
while count<=5:
    if count==3:
        break
    print("hello world")
    count=count+1
   
```

*continue statement:continue statement is skip the current iteration

```python
for i in range(1,11,1):
    if i==5:
        continue
    print(i)
```

```python
count = 1

while count <= 5:

    if count == 3:
        count = count + 1
        continue

    print("hello world")
    count = count + 1
```

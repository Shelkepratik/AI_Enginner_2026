#### 3-The .format() Method (Python 2.7+)

```python
# Positional arguments
print("The {} is flying over the {}.".format("bird", "castle"))

# Indexed arguments
print("{1} text goes here, then {0} text.".format("second", "first")) 
# Output: first text goes here, then second text.

# Keyword arguments
print("{user} logged in from {ip}.".format(user="Bob", ip="192.168.1.1"))

# Modifiers still work similarly
print("Total: {:.2f}".format(12.3456))  # Total: 12.35
```

#### 4- Legacy % Operator (C-Style)

```python
# %s for string, %d for integer, %f for float
print("Hello %s, you scored %d points." % ("Charlie", 95))

# Float precision rounding
print("Pi is roughly %.3f" % 3.14159)  # Pi is roughly 3.142
```

#### without "with-
- risky
- if exception occurs before close()
- file may get locked,blocking for other programs
- data may not be fully written in disk

```python
fobj7=open("demo1.txt","r")
result=fobj7.read()
print(result)
fobj7.close()
```

### using "with" keywoed:
- safe
- file closing automatically when block exits.
- no need to call close() manually
- simple readable code
- works even with exception rasied

```python
with open("demo1.txt","r") as fobj8:
    result=fobj8.read()
    print(result)
```

```python
msg3="""hello world
welcome to Agentic Engineer Course"""
```

```python
with open("demo3.txt","w") as fobj9:
    fobj9.write(msg3)
```

```python
with open("demo3.txt","r") as fobj10:
    result=fobj10.read()
    print(result)
```

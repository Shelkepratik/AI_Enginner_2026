```python
class Patient:
    def __init__(self,patient_name,disease):
        self.patient_name=patient_name
        self.disease=disease
    def display(self):
        return f"The patient name is {self.patient_name} and disease is {self.disease}"
p1=Patient("pratik","XYZ")
p2=Patient("om","abc")
print(p1.display())
print(p2.display())
```

Create a class named Converter with a static method km_to_miles(km) that returns km × 0.621. Call it using the class name without creating any object, for 10 km and 42 km.

```python
class Converter:
    @staticmethod
    def km_to_miles(km):
        return km*0.621
print(Converter.km_to_miles(10)) 
print(Converter.km_to_miles(42))    
```

Create a class named Validator with a static method is_valid_age(age) that returns True if the age is between 1 and 120, otherwise False. Call it for the values 25, 0 and 150.

```python
class validator:
    def is_valid_age(age):
        if age>1 and age<120:
            return True
        else:
            return False
print(validator.is_valid_age(25))
print(validator.is_valid_age(0))
print(validator.is_valid_age(150))
```

Create a class named Shop with a static variable shop_name = "Fresh Mart" and instance variables item and price. Add an instance method show_item() and a static method shop_timing() that prints "Open from 9 AM to 9 PM". Call both methods correctly.

```python

class Shop:
    shop_name = "Fresh Mart"  # Static variable

    def __init__(self, item, price):
        self.item = item      # Instance variable
        self.price = price    # Instance variable

    def show_item(self):       # Instance method
        print("Shop Name:", Shop.shop_name)
        print("Item:", self.item)
        print("Price:", self.price)

    @staticmethod
    def shop_timing():         # Static method
        print("Open from 9 AM to 9 PM")


# Create an object
s1 = Shop("Milk", 60)

# Call the instance method using the object
s1.show_item()

# Call the static method using the class name
Shop.shop_timing()
```

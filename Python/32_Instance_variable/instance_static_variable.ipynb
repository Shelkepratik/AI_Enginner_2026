### Instance variable and static variable

1)Instance variable:variable defined inside the init method using self keyword is known as instance variable
- for each object it has unique own value
- obj_name.instance_variable_name
- example:username,address

```python
class Employee:
    def __init__(self,emp_name,dep):
        #instance variable
        #here.instamce variable:emp_name,sep
        self.emp_name=emp_name
        self.dep=dep
emp1=Employee("pratik","IT")
print(emp1)
print(emp1.emp_name)
print(emp1.dep)
```

2)Static variable:-static variable are inside the class shared among all objects
- shared same value access the object
- class_name.static_variable
- ex:ifsc code

```python
class Employee:
    #static variable
    company_name="TCS"
    def __init__(self,emp_name,dep):
        #instance variable
        #here,instamce variable:emp_name,sep
        self.emp_name=emp_name
        self.dep=dep
emp1=Employee("pratik","IT")
emp2=Employee("om","cs")
print("Empoyee name:",emp1.emp_name)
print("Department name:",emp1.dep)
print("company name:",Employee.company_name)

print("=====================")

print("Empoyee name:",emp2.emp_name)
print("Department name:",emp2.dep)
print("company name:",Employee.company_name)
```

### Instance method and static method

1)instance method:-
- first parameter is self
- access to instance variable
-access to static variable:class_name.static variable
- how to access:
    object_name.methode()
- usecase-act on specific data

```python
class Employee:
    company_name="TCS"
    def __init__(self,emp_name,dep):#instance variable
        self.emp_name=emp_name
        self.dep=dep
    def display_info(self):#instance method
        print("The Empoyee name:",self.emp_name)#access the instance variable
        print("empoyee deparment:",self.dep)
        print("Company name:",Employee.company_name)#access the static variable
emp1=Employee("pratik","IT")
emp1.display_info()
```

2)static method:
- static method is an utility function,conceptually belonges to class but cannot access specific data of an object
- instance variable not able to access
- no self keyword use.
- we have do not access the instance variable.beacuse no object refernce
- access static varible 
- how to access?-class_name.method_name
- usecase-utility m/helper function class level


 in static method define by:
 - mention @staticmethod  above def keyword

 



when to use static method:
- helper/utility function-method is conceptually belong to class but does not requied specific object data

```python
class Employee:
    company_name="TCS"
    def __init__(self,emp_name,dep):#instance variable
        self.emp_name=emp_name
        self.dep=dep

    def display_info(self):#instance method
        print("The Empoyee name:",self.emp_name)#access the instance variable
        print("empoyee deparment:",self.dep)
        print("Company name:",Employee.company_name)#access the static variable

    @staticmethod
    def display_info_new(address):
        print("The company name:",Employee.company_name)
        print("and company address is:",address)

emp1=Employee("pratik","IT")
emp1.display_info()
Employee.display_info_new("pune")
```

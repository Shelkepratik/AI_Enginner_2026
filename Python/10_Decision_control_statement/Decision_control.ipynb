#### Decision control statement

1-if control statement=In if block,else block is optional
-if the if block condition is true then all the statement inside the if block will execute.
-if the if block condition is flase  then all the statement inside the if block will  not execute.
#### syntax=
if condition:
    statement 1
    statement 2

```python
age=int(input("Enter your age"))
print("The age of person: ",age)
if age>=18:
    print("you are eligible for voting")
print("program executed successfully")
```

2-if-else block=The if-else statement is used to execute one block of code when a condition is True and another block when the condition is False. It helps programs make decisions based on different conditions.

```python
age=int(input("Enter your age"))
print(age)
if age>=18:
    print('you are eligible for voting')
else:
    print("you not eligible for voting")
```

3-if elif else block=The if-elif-else statement is used to check multiple conditions one by one. When a condition becomes True, its corresponding block of code is executed. If none of the conditions are True, the else block runs.

```python
age=int(input("Enter your age:"))
if age<10:
    print("you are child")
elif age<18:
    print("you are teenager,but not eligible for voting")
else:
    print("you are eligible for voting")

print("program excuted")
```

4-Nested If-Else Statement
A nested if-else statement is an if-else structure placed inside another if or else block. It is used to check multiple conditions step by step and execute code based on those conditions.

```python
i = 10
if i == 10:
    
    if i < 15:
        print("i is smaller than 15")
        
    if i < 12:
        print("i is smaller than 12 too")
    else:
        print("i is greater than 15")
        
else:
  print("i is not equal to 10")
```

Create a dictionary dict4 = {"mobile": "iphone", "model": "iphone18"}. Make a clone using copy(). Print the ids of both dictionaries, then add a new key "color" to the clone and print both dictionaries.

```python
dict4={"mobile":"iphone","model":"iphone18"}

dict5=dict4.copy()

print(id(dict4),id(dict5))

dict5["color"]="black"#add the new key-value pair in clone

print(dict4)

print(dict5)
```

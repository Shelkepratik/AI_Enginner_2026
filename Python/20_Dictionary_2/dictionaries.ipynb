## Basic operations in dictionaries

### 2. conversion

### converting tuple into dictionary

```python
tuple1=(("name","ganesh"),("age",25))
```

```python
print(tuple1)
```

```python
print(tuple1[0])
```

```python
print(tuple1[1])
```

```python
print(tuple1[0][0])
print(tuple1[0][1])
```

```python
print(tuple1[1][0])
print(tuple1[1][1])
```

```python
dict1=dict(tuple1)
```

```python
print(tuple1)
```

```python
print(dict1)
```

### converting list into dictionary

```python
list1=[("name","ganesh"),("age",25)]
```

```python
dict2=dict(list1)
```

```python
print(dict2)
```

### converting set of elements as key and default value as dictionary

```python
set1={"num1","num2","num3"}

dict3={key:100 for key in set1}

print(dict3)
```

### converting range as keys and values as default to dictionary

```python
range1=range(1,6,1)

dict4={key:10 for key in range1}

print(dict4)
```

### 3. Aliasing

```python
dict1={
    "mobile":"iphone",
    "model":"iphone18"
}
```

```python
dict2=dict1
```

```python
print(id(dict1))
print(id(dict2))
```

```python
dict1["RAM"]="16GB"
```

```python
print(dict1)
```

```python
print(dict2)
```

### 4. Cloning

```python
dict4={
    "mobile":"iphone",
    "model":"iphone18"
}
```

```python
dict5=dict4.copy()
```

```python
print(id(dict4))
print(id(dict5))
```

```python
print(dict5)
print(dict4)
```

```python
dict4["RAM"]="16GB"
```

```python
print(dict5)
print(dict4)
```

# Python OOP: Polymorphism 

**Polymorphism** means “many forms.” In OOP, the same method name, function, or operator can perform different actions depending on the object or data type.

## Definition

**Polymorphism** means “many forms.” In OOP, the same method name, function, or operator can perform different actions depending on the object or data type.

## Important points for interviews

- One interface can represent different underlying behaviors.
- Method overriding is a common runtime form of polymorphism.
- Operator overloading lets operators work with user-defined objects using special methods such as `__add__`.
- Python supports duck typing: if an object provides the needed behavior, it can be used.
- Polymorphism makes code flexible, reusable, and easier to extend.
- Different classes do not always need to share a parent class to be used polymorphically in Python.
- Common examples include `len()` working with strings, lists, and dictionaries.
- Method overloading by defining multiple same-named methods with different signatures is not supported in the traditional way; later definitions replace earlier ones. Default arguments or `*args` can be used instead.

## Examples

Each example includes a short explanation and runnable Python code.

### Example 1: Method overriding

A child class provides its own version of a method inherited from the parent class.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

Animal().sound()
Dog().sound()
```

### Example 2: Same method name, different classes

The same call produces behavior based on the object type.

```python
class Cat:
    def speak(self):
        return "Meow"

class Cow:
    def speak(self):
        return "Moo"

for animal in (Cat(), Cow()):
    print(animal.speak())
```

### Example 3: Polymorphism with a common function

A single function can work with different objects as long as they provide the required method.

```python
class Circle:
    def draw(self):
        print("Drawing a circle")

class Square:
    def draw(self):
        print("Drawing a square")

def display_shape(shape):
    shape.draw()

display_shape(Circle())
display_shape(Square())
```

### Example 4: Operator overloading

The + operator behaves differently for numbers and strings; custom classes can define operator behavior using special methods.

```python
print(10 + 20)
print("Hello " + "Python")

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2
print(p3.x, p3.y)
```

### Example 5: Duck typing

Python often uses an object's available behavior rather than requiring a specific class or inheritance relationship.

```python
class Laptop:
    def start(self):
        print("Laptop starts")

class Car:
    def start(self):
        print("Car starts")

def start_device(device):
    device.start()

start_device(Laptop())
start_device(Car())
```

### Example 6: Polymorphism with built-in functions

The same built-in function can operate on different data types, with behavior appropriate to each type.

```python
print(len("Python"))
print(len([10, 20, 30]))
print(len({"name": "Asha", "age": 21}))
```

## Interview questions and short answers

**Q1. What is polymorphism?**

It means one interface or method name can have different behaviors for different objects.

**Q2. What is method overriding?**

A child class defines a method with the same name as a parent method to provide specialized behavior.

**Q3. What is operator overloading?**

Giving an operator a behavior for a user-defined class through special methods such as `__add__`.

**Q4. What is duck typing?**

Python focuses on whether an object supports the required operation, rather than its exact class.

**Q5. Does Python support traditional method overloading?**

Not directly. Defining the same method name multiple times replaces the earlier definition; use default values or `*args` for flexible arguments.

**Q6. Why is polymorphism useful?**

It lets one piece of code work with different object types, improving extensibility and reducing duplicate code.

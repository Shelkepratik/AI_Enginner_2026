### *Class and Object

- class is a blueprint
- class contain Attribute and methods

- class define the attribute of object

2-Object-object is entity that follow class rules.
- object is instance of class
- follow rules of class

```python
class vehical:
    def __init__(self):
        self.color="red"
        self.top_speed=160
        self.milege=15
    def dispaly_info(self):
        print("color os vehical: ",self.color)
        print("milage is: ",self.milege)
        print("top speed is: ",self.top_speed)
vehical1=vehical()
vehical1.dispaly_info()
```

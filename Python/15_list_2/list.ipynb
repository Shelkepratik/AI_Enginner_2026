## List all method




| Method      | Syntax                  | Example            | Description                                                   |
| ----------- | ----------------------- | ------------------ | ------------------------------------------------------------- |
| `append()`  | `list.append(x)`        | `a.append(10)`     | Adds one element at the end.                                  |
| `extend()`  | `list.extend(iterable)` | `a.extend([4, 5])` | Adds multiple elements to the end.                            |
| `insert()`  | `list.insert(i, x)`     | `a.insert(1, 20)`  | Inserts an element at a specified index.                      |
| `remove()`  | `list.remove(x)`        | `a.remove(10)`     | Removes the first matching element.                           |
| `pop()`     | `list.pop(i)`           | `a.pop()`          | Removes and returns an element; by default, the last element. |
| `clear()`   | `list.clear()`          | `a.clear()`        | Removes all elements.                                         |
| `index()`   | `list.index(x)`         | `a.index(20)`      | Returns the index of the first matching element.              |
| `count()`   | `list.count(x)`         | `a.count(10)`      | Counts occurrences of an element.                             |
| `sort()`    | `list.sort()`           | `a.sort()`         | Sorts the list in ascending order by default.                 |
| `reverse()` | `list.reverse()`        | `a.reverse()`      | Reverses the order of elements.it change in original list                             |
| `copy()`    | `list.copy()`           | `b = a.copy()`     | Creates a shallow copy of the list.                           |

Note: Assume `a = [10, 20, 10, 30]` unless otherwise specified. The examples are independent, so each method starts with the original list.

```python
list1=[1,2,3,4,5]
print(list1.reverse())
print(list1)
```

## function



| Function      | Syntax                       | Example                         | Description                                             |
| ------------- | ---------------------------- | ------------------------------- | ------------------------------------------------------- |
| `len()`       | `len(list)`                  | `len([10, 20, 30])`             | Returns the number of elements.                         |
| `max()`       | `max(list)`                  | `max([10, 20, 30])`             | Returns the largest element.                            |
| `min()`       | `min(list)`                  | `min([10, 20, 30])`             | Returns the smallest element.                           |
| `sum()`       | `sum(list)`                  | `sum([10, 20, 30])`             | Returns the sum of numeric elements.                    |
| `sorted()`    | `sorted(list)`               | `sorted([30, 10, 20])`          | Returns a new sorted list.it create a new list                              |
| `list()`      | `list(iterable)`             | `list("abc")`                   | Converts an iterable into a list.                       |
| `any()`       | `any(list)`                  | `any([0, 0, 5])`                | Returns `True` if at least one element is truthy.       |
| `all()`       | `all(list)`                  | `all([1, 2, 3])`                | Returns `True` if all elements are truthy.              |
| `enumerate()` | `enumerate(list)`            | `list(enumerate(["A", "B"]))`   | Returns index-value pairs.                              |
| `reversed()`  | `reversed(list)`             | `list(reversed([1, 2, 3]))`     | Returns a reverse iterator.                             |
| `filter()`    | `filter(function, iterable)` | `list(filter(None, [0, 1, 2]))` | Selects elements for which the function returns `True`. |
| `map()`       | `map(function, iterable)`    | `list(map(str, [1, 2, 3]))`     | Applies a function to every element.                    |

Note: These are Python built-in functions that can be used with lists; they are not list methods.

```python
list1=[1,2,3,4,5]
result=reversed(list1)
print(list(result))
```

## property



| Property                      | Syntax                  | Example                  | Description                                       |
| ----------------------------- | ----------------------- | ------------------------ | ------------------------------------------------- |
| Ordered                       | `list`                  | `a = [10, 20, 30]`       | Elements maintain their order.                    |
| Mutable                       | `list[index] = value`   | `a[0] = 100`             | Elements can be changed after creation.           |
| Allows duplicates             | `list`                  | `a = [10, 10, 20]`       | A list can contain duplicate elements.            |
| Allows different data types   | `list`                  | `a = [10, "Hello", 2.5]` | A list can store different types of values.       |
| Indexed                       | `list[index]`           | `a[0]`                   | Elements can be accessed using indexes.           |
| Supports negative indexing    | `list[-index]`          | `a[-1]`                  | Accesses elements from the end.                   |
| Dynamic size                  | `append()` / `remove()` | `a.append(40)`           | The list can grow or shrink.                      |
| Allows nested lists           | `list` inside `list`    | `a = [[1, 2], [3, 4]]`   | A list can contain other lists.                   |
| Created using square brackets | `[ ]`                   | `a = [1, 2, 3]`          | Lists are commonly created using square brackets. |
| Empty list supported          | `[]`                    | `a = []`                 | A list can contain zero elements.                 |

Remember: Python lists are ordered, mutable, and allow duplicate values and different data types.

Write a function print_details(**kwargs) that prints kwargs and its type. Call it with name="rahul", age=20, city="pune".

Using the print_details(**kwargs) function from Q7, add a for loop with items() to print each key and value on a separate line.

```python
def print_details(**kwargs):
    print(kwargs)
    for k,v in kwargs.items():
        print(k,":",v)   
print_details(name="rahul",age=20,city="pune")
```

Write a function student_info(**kwargs) that prints only the keys passed to it using keys(). Call it with name="sneha", roll_no=51, course="python".

```python
def student_info(**kwargs):
    for k in kwargs.keys():
        print(k)
student_info(name="sneha",roll_no=51,course="python")
```

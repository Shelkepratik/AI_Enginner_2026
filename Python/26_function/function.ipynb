*Type of argument:

1)positional argument:this is the most common type.

we pass the value on the basic of position and assign it to the parameters 

 we make sure that passing value in the correct position

```python
def greet(name,age):
    print("my name is ",name,"and age is ",age)

greet("pratik",23)
```

2)keyword argument:we pass the arguments in any order

we match the keywords with the parameters used in the function



```python
def greet(name,age):
    print("my name is ",name,"and age is ",age)

greet(name="pratik",age=20)
greet(age=20,name="pratik")
```

3)Default argument:we can provide default values to a parameters.and if we donot pass a value then it will consider this default value

without deafult argument

```python
def greet(name,age):
    print("my name is: ",name,"and age is ",age)
greet("pratik")#TypeError
```

with default argument

```python
def greet(name="pratik",age=23):
    print("my name is: ",name,"and age is: ",age)
greet("om")
greet()
greet(age=22)
```

## *args and **kwargs

## 4)variable length argument:we receive these arguments in *args

- A variable-length argument allows a function to accept any number of arguments.

- it taks *args as a tuple


```python
def calculate_sum(*args):
    print(args)
    print(type(args))
    return sum(args)
result=calculate_sum(1,2,3)
print("The result is: ",result)
```

```python
result1=calculate_sum(1,2,3,4,5,6,7)
print(result1)
```

- args-tuple
- kwargs-dictionary

5)variable length keyword argument(**kwargs):-using variable length kwyword argument.we can pass many keyword argument to the function

- **kwargs can take as many "kwword" argument 

- inside function kwargs act as a dictionary.

- the kwword we pass we taje as a key-value pair

```python
def print_details(**kwargs):
    print(kwargs)
    print(type(kwargs))
print_details(name="rahul",age=23,city="pune")
```

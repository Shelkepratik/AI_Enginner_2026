## Recursion

- recursion programming techniques in which one function calls itself again and again and function call exits in the function definition

```python
def hello():
    print("hello")
    hello()
result=hello()
print(result)
```

```python
import sys
print(sys.getrecursionlimit())
```

```python
def count_up(start,stop):
    if start>stop:
        return start
    print(start)
    return count_up(start+1,stop)

result=count_up(1,5)
print(result)
```

```python
def fact(num):
    if num==1:
        return 1
    return num*fact(num-1)
n=5
print(fact(n))
```

Write a recursive function countdown(n) that prints numbers from n down to 1, and then prints "Lift off!". Call it with n = 5.

```python
def countdown(n):
    if n==0:
        return
    print(n)
    return countdown(n-1)

countdown(5)
print("Lift off!")
```

Write a recursive function sum_upto(n) that returns the sum of all numbers from 1 to n. Test it with n = 10 (expected answer: 55).

```python
def sum_upto(n):
    if n==0:
        return 0
    return n + sum_upto(n-1)
    

sum_upto(10)
```

Write a recursive function factorial(n) that returns the factorial of a number. Test it with n = 5 (expected answer: 120).

```python
def factorial(n):
    if n==1:
        return 1 

    return n * factorial(n-1)

print(factorial(5))
```

Write a recursive function print_even(start, stop) that prints only the even numbers between start and stop. Test it with (1, 20).

```python
def print_even(start,stop):
    if start>stop:#Base case
        return

    if start % 2==0:
        print(start)

    print_even(start+1,stop)#recursive call

print_even(1,20)

```

What is bytecode in python 

- जेव्हा आपण Python मध्ये code लिहितो, तेव्हा Python तो code bytecode मध्ये convert करतो. त्यानंतर Python Virtual Machine (PVM) bytecode execute करते आणि output मिळतो.
- Python source code ला Python bytecode मध्ये convert करतो. Bytecode हा original code पेक्षा lower-level असतो, पण तो directly machine code नसतो.
- Bytecode is an intermediate, lower-level representation of Python source code, but it is not machine code.

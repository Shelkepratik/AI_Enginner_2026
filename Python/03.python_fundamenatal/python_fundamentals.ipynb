## What are Identifiers?
#### An identifier is a name used to identify a variable, function,class, module, or other object

```python
#Variable -> age = 25
```

```python
#Function -> def sum():
```

```python
#Class -> class Employee:
```

```python
#Module -> import math
```

## Rules for Creating Identifiers

#### 1) Start with alphabet or underscore
#### 2) Followed by letters, digits, or underscores
#### 3) Keywords cannot be used as identifiers
#### 4)identifies is case sensitive
#### 5)identifier canot be contain the special symbol,space

```python
1a=10
```

```python
@a=10
```

```python
a12=10
```

```python
_a12=20
```

```python
print(a12)
```

```python
print(_a12)
```

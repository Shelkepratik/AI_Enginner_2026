## Definition

**Abstraction** is an OOP concept that hides implementation details and exposes only the essential features. In Python, abstraction is commonly implemented using abstract classes and abstract methods from the `abc` module.

## Important points for interviews

- Focuses on what an object does, not how it does it.
- An abstract class is created by inheriting from `ABC`.
- An abstract method is marked with `@abstractmethod`.
- A class with unimplemented abstract methods cannot be instantiated.
- A concrete child class must implement all inherited abstract methods.
- An abstract class can contain normal (concrete) methods, constructors, and data.
- Abstraction helps define a common contract for related classes.
- Python does not enforce abstraction through a special keyword; it uses the `abc` module.
- Do not confuse abstraction (hiding implementation complexity) with encapsulation (restricting/directing access to data).

## Examples

Each example includes a short explanation and runnable Python code.

### Example 1: Abstract class with an abstract method

An abstract class defines a common interface. A child class must implement its abstract method before it can be instantiated.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = Rectangle(5, 3)
print("Area:", r.area())
```

### Example 2: Different subclasses implement the same abstract method

Every payment type provides its own implementation of the required pay() method.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class UPI(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")

class Card(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Card")

UPI().pay(500)
Card().pay(1200)
```

### Example 3: Abstract class with multiple abstract methods

A subclass must implement all abstract methods to become concrete.

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass

    @abstractmethod
    def show_role(self):
        pass

class Developer(Employee):
    def calculate_salary(self):
        return 60000

    def show_role(self):
        return "Python Developer"

dev = Developer()
print(dev.show_role())
print("Salary:", dev.calculate_salary())
```

### Example 4: Abstract class with a concrete method

An abstract class can contain both abstract methods and already-implemented methods.

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    def plug_in(self):
        print("Appliance is plugged in")

class Fan(Appliance):
    def turn_on(self):
        print("Fan is running")

fan = Fan()
fan.plug_in()
fan.turn_on()
```

### Example 5: Cannot instantiate an abstract class

- Python prevents creating an object of a class while abstract methods remain unimplemented.
- This code demonstrates abstraction using an abstract class. It shows that we cannot create an object of an abstract class until its abstract methods are implemented.


```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

try:
    vehicle = Vehicle()
except TypeError as error:
    print("Error:", error)

class Bike(Vehicle):
    def start(self):
        print("Bike started")

Bike().start()
```

### Example 6: Abstract class as a common blueprint

Different report types follow the same required generate() contract.

```python
from abc import ABC, abstractmethod

class Report(ABC):
    @abstractmethod
    def generate(self):
        pass

class PDFReport(Report):
    def generate(self):
        return "Generating PDF report"

class ExcelReport(Report):
    def generate(self):
        return "Generating Excel report"

for report in (PDFReport(), ExcelReport()):
    print(report.generate())
```

## Interview questions and short answers

**Q1. What is abstraction?**

It hides internal implementation details and shows only the essential operations to the user.

**Q2. How do you achieve abstraction in Python?**

Using the `ABC` class and `@abstractmethod` decorator from the `abc` module.

**Q3. Can we create an object of an abstract class?**

No, not while it has abstract methods that are not implemented.

**Q4. Can an abstract class have a normal method?**

Yes. It can contain both abstract and concrete methods.

**Q5. What happens if a child class does not implement an abstract method?**

The child class remains abstract and cannot be instantiated.

**Q6. Abstraction vs encapsulation?**

Abstraction hides complexity; encapsulation bundles data and methods and controls access to internal state.

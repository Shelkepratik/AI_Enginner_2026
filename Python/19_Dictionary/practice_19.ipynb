Create a dictionary phone = {"brand": "Samsung", "model": "S24", "color": "black"}. Use enumerate() with items() to print the index, key and value of each pair

```python
phone={"brand":"samsung",
       "model":"s24",
       "color":"brack"
       }
for index,(key,value) in enumerate(phone.items()):
    print(index,key,value)
```

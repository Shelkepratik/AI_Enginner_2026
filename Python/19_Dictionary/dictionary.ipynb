*dictionary is also one of the container data type in python

dict is a collection of key-value pair

*key value pairs are seperated by comma and enclosed in curly brackets

```python
dict1={"name":"Ganesh",
       "age":23,
        "course":"MCA"
       }
print(dict1)
```

*different keys may have same value

```python
dict2={"name":"ganesh",
       "roll_no":25,
       "age":25}
print(dict2)
```

*keys should be unique but if we have same key with different value,it will consider latest key,value pair

```python
dict1={
    "name":"ganesh",
    "roll_no":25,
    "name":"pratik"
}
print(dict1)
```

*Accessing dictionary element-

-to print/return the entire dictionary we use dictionary name

```python
dict1={"name":"ganesh",
       "age":25}
print(dict1)
print(type(dict1))
```

-using keys we can access/print the values

```python
print(dict1["name"])
print(dict1["age"])
```

*Looping in dictionaries:-if we wnat to process each element in the dictionary,we need to iterate the dictionary and we do it using from loop

1)for loop-use items()

```python
dict1={"name":"pratik",
       "age":25}
for key,value in dict1.items():
    print(key,value)
```

2)for loop-keys()-To iterate keys

```python
dict1={"car":"Tata",
       "model":"nexon"}
print(dict1)
for keys in dict1.keys():
    print(keys)
```

3)for loop-values()-To iterate values

```python
dict1={"name":"pratik","age":23}
print(dict1)
for values in dict1.values():
    print(values)
```

4)enumerate:

```python
for i,(k,v) in enumerate(dict1.items()):
    print(i,k,v)
```

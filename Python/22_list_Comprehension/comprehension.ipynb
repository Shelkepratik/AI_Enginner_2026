## Comprehension:


comprehension are concise,redable way to create list,set and dictionary by iterating over a sequence and optionally applying a condition

all in a single line of code

1)create easily

2)iterate  and assign result to container data type

3)not for tuple-tuple are immutable

### 1) list comprehension

list comprehention create a new list by iterating over a sequence and applying an epression to each item



*syntax=[expression for item in iterable(sequence  if condition)]

iterable-list,set,dictionary

if condtion-is optional

1)create a list having squre of number from 1 t0 5

o/p=[1,4,9,16,25]

```python
for i in range(1,6):
    print(i ** 2,end=" ")
```

```python
list1=[i ** 2 for i in range(1,6,1)]
print(list1)
```

2)converting words into uppercase

```python
animal=["cat","dog","fish","Tiger"]
new_list=[i.upper() for  i in animal ]
print(new_list)
```

```python
new_list1=[i.upper() for i in animal if len(i)<=3]
print(new_list1)
```

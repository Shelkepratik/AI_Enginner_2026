Create a list animals = ["cat", "dog", "fish", "tiger"]. Using list comprehension, create a new list with all the animal names in uppercase

```python
animals=["cat","dog","fish","tiger"]

new_list=[i.upper() for i in animals]

print(new_list)
```

Using the animals list from Q5, create a new list containing only the animal names whose length is 4 or less, in uppercase. Solve it once using a for loop with append() and once using list comprehension.

```python
new_list=[i.upper() for i in animals if len(i)<=4]

print(new_list)

```

```python
new_list=[]

for i in animals:
    if len(i)<=4:
        new_list.append(i.upper())

print(new_list)
```

Create a list numbers = [5, 12, 3, 18, 7, 25, 10]. Using list comprehension, create a new list containing only the numbers greater than 10.

```python
numbers=[5,12,3,18,7,25,10]

new_list=[i for i in numbers if i>10]
print(new_list)
```

Create a list names = ["ravi", "amit", "sneha", "priya"]. Using list comprehension, create a new list containing the length of each name.

```python
names=["ravi","amit","sneha","priya"]

new_list=[len(i) for i in names]

print(new_list)
```

Create a list fruits = ["apple", "banana", "mango", "kiwi"]. Using list comprehension, create a new list containing only the fruits that start with the letter "m" or "k"

```python
fruits = ["apple", "banana", "mango", "kiwi"]

new_fruits = [fruit for fruit in fruits if fruit.startswith("m") or fruit.startswith("k")]

print(new_fruits)
```

Create a list prices = [100, 250, 400, 50]. Using list comprehension, create a new list where each price is increased by 10.

```python
prices = [100, 250, 400, 50]

new_prices = [price + 10 for price in prices]

print(new_prices)
```

Create a class named Student with a private variable __marks set to 0 in the constructor. Add a method update_marks(new_marks) that updates the marks only if the value is between 0 and 100, otherwise prints "Invalid marks". Add a method show_marks() to print the marks. Test it with valid and invalid values.

```python
class Student:
    def __init__(self):
        self.__marks=0
    def update_marks(self,new_marks):
        if new_marks>0 and new_marks<100:
            self.__marks=new_marks
        else:
            print('invalid marks')
    def show_marks(self):
        print("The marks is:",self.__marks)
s1=Student()
s1.update_marks(55)
s1.show_marks()
```

 Using the Student class from Q8, try to print obj.__marks directly from outside the class. Write a comment about the error you get and why.


```python

class Student:
    def __init__(self):
        self.__marks = 0

    def update_marks(self, new_marks):
        if 0 <= new_marks <= 100:
            self.__marks = new_marks
        else:
            print("Invalid marks")

    def show_marks(self):
        print("Marks:", self.__marks)


obj = Student()

obj.update_marks(85)
obj.show_marks()

# Trying to access the private variable directly
print(obj.__marks)

# Error: AttributeError occurs because __marks is a private-style
# attribute and Python uses name mangling to change its internal name.
```

Create a class named Wallet with a private variable _amount = 1000. Outside the class, write obj._amount = 0 and then call a method show_amount() that prints the private amount. Write a comment explaining why the amount shown is still 1000. Then add a method add_money(value) that allows adding only positive values to the wallet.

```python

class Wallet:
    def __init__(self):
        self._amount = 1000

    def show_amount(self):
        print("Wallet Amount:", self._amount)

    def add_money(self, value):
        if value > 0:
            self._amount += value
            print("Money added successfully")
        else:
            print("Please enter a positive amount")


obj = Wallet()

obj._amount = 0
obj.show_amount()

# Note: A single underscore does not prevent direct access.
# Therefore, obj._amount = 0 changes the amount to 0,
# and show_amount() prints 0, not 1000.

obj.add_money(500)
obj.show_amount()

obj.add_money(-200)
obj.show_amount()
```

Create a class named Student with a private variable __marks set to 0 in the constructor. Add a method update_marks(new_marks) that updates the marks only if the value is between 0 and 100, otherwise prints "Invalid marks". Add a method show_marks() to print the marks. Test it with valid and invalid values.

```python
class Student:
    def __init__(self):
        self.__marks=0
    def update_marks(self,new_marks):
        if new_marks>0 and new_marks<100:
            self.__marks=new_marks
        else:
            print('invalid marks')
    def show_marks(self):
        print("The marks is:",self.__marks)
s1=Student()
s1.update_marks(55)
s1.show_marks()
```

Create a class named Wallet with a private variable _amount = 1000. Outside the class, write obj._amount = 0 and then call a method show_amount() that prints the private amount. Write a comment explaining why the amount shown is still 1000. Then add a method add_money(value) that allows adding only positive values to the wallet.

```python

class Wallet:
    def __init__(self):
        self._amount = 1000

    def show_amount(self):
        print("Wallet Amount:", self._amount)

    def add_money(self, value):
        if value > 0:
            self._amount += value
            print("Money added successfully")
        else:
            print("Please enter a positive amount")


obj = Wallet()

obj._amount = 0
obj.show_amount()

# Note: A single underscore does not prevent direct access.
# Therefore, obj._amount = 0 changes the amount to 0,
# and show_amount() prints 0, not 1000.

obj.add_money(500)
obj.show_amount()

obj.add_money(-200)
obj.show_amount()
```

Create a class named Product with a private variable __price. Add a getter get_price() and a setter set_price(new_price) that allows only positive numbers. Try setting the price to 499, -50 and "free", and print the price after each attempt.

```python
class Product:
    def __init__(self):
        self.__price=0

    def get_price(self):
        return self.__price
    def set_price(self,new_price):
        if isinstance(new_price,(int,float)) and not isinstance(new_price,bool) and new_price>0:
            self.__price=new_price
        else:
            print("Invalid price")

p1=Product()
p1.set_price(499)
print("price:",p1.get_price())

p1.set_price(-50)
print("price:",p1.get_price())

p1.set_price("free")
print("price:",p1.get_price())
```

Create a class named Thermostat with a private variable __temperature = 24. Add a getter get_temperature() and a setter set_temperature(value) that allows only values between 16 and 30. Add a method increase() and decrease() that change the temperature by 1 degree using the setter, so the limits are always respected. Test what happens when you call increase() many times.

```python

class Thermostat:
    def __init__(self):
        self.__temperature = 24

    def get_temperature(self):
        return self.__temperature

    def set_temperature(self, value):
        if 16 <= value <= 30:
            self.__temperature = value
        else:
            print("Invalid temperature")

    def increase(self):
        self.set_temperature(self.__temperature + 1)

    def decrease(self):
        self.set_temperature(self.__temperature - 1)


t1 = Thermostat()

print("Initial temperature:", t1.get_temperature())

t1.increase()
print("After increase:", t1.get_temperature())

t1.decrease()
print("After decrease:", t1.get_temperature())

# Test increase many times
for i in range(10):
    t1.increase()

print("Final temperature:", t1.get_temperature())

```

## set property



| Property                   | Syntax               | Example                  | Description                                                         |
| -------------------------- | -------------------- | ------------------------ | ------------------------------------------------------------------- |
| Unordered                  | `set`                | `s = {10, 20, 30}`       | Elements do not have a guaranteed order.                            |
| Mutable                    | `set`                | `s.add(40)`              | Elements can be added or removed.                                   |
| No duplicates              | `set`                | `s = {10, 10, 20}`       | Duplicate elements are automatically removed.                       |
| No indexing                | `set[index]`         | `s[0]`                   | Sets do not support indexing.                                       |
| No slicing                 | `set[start:end]`     | `s[0:2]`                 | Sets do not support slicing.                                        |
| Different data types       | `set`                | `s = {10, "Hello", 2.5}` | Can contain elements of different hashable types.                   |
| Unique elements            | `set`                | `s = {1, 2, 3}`          | Each element appears only once.                                     |
| Created using curly braces | `{ }`                | `s = {1, 2, 3}`          | Curly braces can be used to create a non-empty set.                 |
| Empty set                  | `set()`              | `s = set()`              | An empty set is created using `set()`, not `{}`.                    |
| Supports set operations    | `&`, `\\|`, `-`, `^` | `{1, 2} & {2, 3}`        | Supports intersection, union, difference, and symmetric difference. |
| Elements must be hashable  | `set`                | `s = {[1, 2]}`           | Lists and other unhashable objects cannot be set elements.          |

Remember: A Python set is unordered, mutable, and does not allow duplicate elements. Although a set is mutable, its individual elements must be hashable.

## method


| Function      | Syntax                | Example                       | Description                                                                     |
| ------------- | --------------------- | ----------------------------- | ------------------------------------------------------------------------------- |
| `len()`       | `len(set)`            | `len({10, 20, 30})`           | Returns the number of elements.                                                 |
| `max()`       | `max(set)`            | `max({10, 20, 30})`           | Returns the largest element.                                                    |
| `min()`       | `min(set)`            | `min({10, 20, 30})`           | Returns the smallest element.                                                   |
| `sum()`       | `sum(set)`            | `sum({10, 20, 30})`           | Returns the sum of numeric elements.                                            |
| `sorted()`    | `sorted(set)`         | `sorted({30, 10, 20})`        | Returns a sorted list of elements.                                              |
| `set()`       | `set(iterable)`       | `set([1, 2, 2, 3])`           | Creates a set and removes duplicates.                                           |
| `frozenset()` | `frozenset(iterable)` | `frozenset([1, 2, 3])`        | Creates an immutable set.                                                       |
| `any()`       | `any(set)`            | `any({0, 1, 2})`              | Returns `True` if at least one element is truthy.                               |
| `all()`       | `all(set)`            | `all({1, 2, 3})`              | Returns `True` if all elements are truthy.                                      |
| `enumerate()` | `enumerate(set)`      | `list(enumerate({"A", "B"}))` | Returns index-element pairs in the set's iteration order.                       |
| `reversed()`  | `reversed(set)`       | `reversed({1, 2, 3})`         | Not supported for sets because sets do not maintain a defined positional order. |

Note: These are Python built-in functions that work with sets. They are different from set methods such as `add()`, `remove()`, `union()`, and `intersection()`.

## method



| Method                          | Syntax                                   | Example                                 | Description                                                                 |
| ------------------------------- | ---------------------------------------- | --------------------------------------- | --------------------------------------------------------------------------- |
| `add()`                         | `set.add(x)`                             | `s.add(10)`                             | Adds one element to the set.                                                |
| `update()`                      | `set.update(iterable)`                   | `s.update([10, 20])`                    | Adds multiple elements to the set.                                          |
| `remove()`                      | `set.remove(x)`                          | `s.remove(10)`                          | Removes an element; raises an error if it does not exist.                   |
| `discard()`                     | `set.discard(x)`                         | `s.discard(10)`                         | Removes an element without an error if it is absent.                        |
| `pop()`                         | `set.pop()`                              | `s.pop()`                               | Removes and returns an arbitrary element.                                   |
| `clear()`                       | `set.clear()`                            | `s.clear()`                             | Removes all elements.                                                       |
| `copy()`                        | `set.copy()`                             | `s2 = s.copy()`                         | Creates a shallow copy of the set.                                          |
| `union()`                       | `set.union(other)`                       | `{1, 2}.union({2, 3})`                  | Returns a set containing elements from both sets.                           |
| `intersection()`                | `set.intersection(other)`                | `{1, 2}.intersection({2, 3})`           | Returns common elements.                                                    |
| `difference()`                  | `set.difference(other)`                  | `{1, 2}.difference({2, 3})`             | Returns elements present in the first set but not the second.               |
| `symmetric_difference()`        | `set.symmetric_difference(other)`        | `{1, 2}.symmetric_difference({2, 3})`   | Returns elements present in either set, but not both.                       |
| `intersection_update()`         | `set.intersection_update(other)`         | `s.intersection_update({2, 3})`         | Updates the original set with common elements.                              |
| `difference_update()`           | `set.difference_update(other)`           | `s.difference_update({2, 3})`           | Removes elements found in the other set.                                    |
| `symmetric_difference_update()` | `set.symmetric_difference_update(other)` | `s.symmetric_difference_update({2, 3})` | Updates the original set with elements present in either set, but not both. |
| `issubset()`                    | `set.issubset(other)`                    | `{1, 2}.issubset({1, 2, 3})`            | Checks whether all elements belong to the other set.                        |
| `issuperset()`                  | `set.issuperset(other)`                  | `{1, 2, 3}.issuperset({1, 2})`          | Checks whether the set contains all elements of the other set.              |
| `isdisjoint()`                  | `set.isdisjoint(other)`                  | `{1, 2}.isdisjoint({3, 4})`             | Returns `True` if the sets have no common elements.                         |

Note: Assume `s = {1, 2, 3}` for examples using `s`, unless otherwise specified. Examples are independent.

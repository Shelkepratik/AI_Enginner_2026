#### *set method

1.add():-using add() method we can add the new element in the set

```python
set1={10,20,30}
print(set1)
set1.add(40)
print(set1)
```

2.update-using this method we can add elements of one set into the another set

```python
set1={10,20,30}
set2={40,50}
set1.update(set2)
print(set1)
```

```python
set1.update({90,80})
print(set1)
```

3.remove()-using remove() method  we can remove paricular element from the set.

```python
set1={10,20,30,40,50}
set1.remove(30)
print(set1)
```

```python
#remove those element not present in the set
set1.remove(80)  #return the error
```

4.discard()-using discard method,we can remove paricular element from the set.

```python
set1={10,20,30,40,50}
set1.discard(30) 
print(set1)
```

```python
#remove those element not present in the set
set1.discard(90)#do return the error
```

remove() vs discard() when we try to remove element from the set which is not available

-using remove method we get keyError

-using discard method we not get error

5.clear()-using clear() method we can delete all the element from the set,it's create empty set

```python
set1={10,20,30,40,50}
set1.clear()
print(set1)
```

6.copy()-it create copy of set

```python
set1={10,20,30,40,50}
set2=set1.copy()
print(set2)
print(set1)

print(id(set1),id(set2))
```

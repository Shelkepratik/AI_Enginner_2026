```python
set1={10,20,30}
set2={40,50}
set1.update(set2)
print(set1)
```

Create a set {10, 20, 30}. Make a copy of it using copy(). Print both sets and their id() values. Are the ids the same or different?

```python
set1={10,20,30}
set2=set1.copy()
print(set1)
print(set2)
print(id(set1),id(set2))
```

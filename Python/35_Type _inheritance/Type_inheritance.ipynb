## Type of inherritance:

### single inheritance

```python
class phone:
    def __init__(self,price,brand,camera):
        print("Inside parent class constructor")
        self.price=price
        self.brand=brand
        self.camera=camera

    def buy(self):
        print("Buying a phone")
        print("we are in the parent class's instance method")


class Smartphone(phone):
    def __init__(self,price,brand,camera):
        print("we inside the child class constructor")
        super().__init__(price,brand,camera)
    def display(self):
        print("we are in the smartphone instance method")

obj1=Smartphone(100000,"apple","20px")
obj1.buy()
obj1.display()
print(obj1.price)
print(obj1.brand)
print(obj1.camera)
```

### multilevel Inheritance

```python
class product:
    def review(self):
        print("Product customer review")

class phone(product):
    def __init__(self,price,brand,camera):
        print("inside phone constructor")
        self.price=price
        self.brand=brand
        self.camera=camera

    def buy(self):
        print("Buying a phone")
        print("we are in the phone class's instance method")


class smartphone(phone):
    pass

obj1=smartphone(100000,"apple","20px")
obj1.buy()
print(obj1.price)
print(obj1.brand)
print(obj1.camera)
obj1.review()
        
```

```python
class Grandfather:
    def __init__(self, family_name):
        self.family_name = family_name
        print("Grandfather constructor called")

    def show_family(self):
        print("Family Name:", self.family_name)

    def show_role(self):
        print("I am Grandfather")


class Father(Grandfather):
    def __init__(self, family_name, father_name):
        super().__init__(family_name)
        self.father_name = father_name
        print("Father constructor called")

    def show_father(self):
        print("Father Name:", self.father_name)

    def show_role(self):
        print("I am Father")


class Child(Father):
    def __init__(self, family_name, father_name, child_name):
        super().__init__(family_name, father_name)
        self.child_name = child_name
        print("Child constructor called")

    def show_child(self):
        print("Child Name:", self.child_name)

    def show_role(self):
        super().show_role()
        print("I am Child")


# Create object
c = Child("Shelke", "Anandrao", "Pratik")

print("\n--- Calling Methods ---")
c.show_family()
c.show_father()
c.show_child()

print("\n--- Method Overriding and super() ---")
c.show_role()

print("\n--- Accessing Attributes ---")
print(c.family_name)
print(c.father_name)
print(c.child_name)

print("\n--- MRO ---")
print([cls.__name__ for cls in Child.mro()])

print("\n--- isinstance() ---")
print(isinstance(c, Child))
print(isinstance(c, Father))
print(isinstance(c, Grandfather))
```

### Multiple inheritance

```python
class phone:
    def __init__(self,price,brand,camera):
        print("inside the phone constructor")
        self.price=price
        self.brand=brand
        self.camera=camera
    def buy(self):
        print("price is:",self.price)
        print("Buying a phone")
        print("we are in the phone class instance methode")

class product:
    def __init__(self,price,brand,camera):
        print("inside the product constructor")
        self.price=price
        self.brand=brand
        self.camera=camera
    def review(self):
        print("product customer review")

class smartphone(phone,product):
    pass



obj1=smartphone(10000,"apple","13px")
obj1.buy()
obj1.review()
#find the all class
print([cls.__name__ for cls in smartphone.mro()])

```

```python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
        print("Father constructor called")

    def show_father(self):
        print("Father Name:", self.father_name)

    def show_skills(self):
        print("Father: Driving")


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
        print("Mother constructor called")

    def show_mother(self):
        print("Mother Name:", self.mother_name)

    def show_cooking(self):
        print("Mother: Cooking")

    def show_skills(self):
        print("Mother: Teaching")


class Child(Father, Mother):
    def __init__(self, father_name, mother_name, child_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.child_name = child_name
        print("Child constructor called")

    def show_child(self):
        print("Child Name:", self.child_name)

    def show_skills(self):
        super().show_skills()
        print("Child: Coding")


# Create object
c = Child("Anandrao", "Pratibha", "Pratik")

print("\n--- Calling Methods ---")
c.show_father()
c.show_mother()
c.show_child()
c.show_cooking()

print("\n--- Method Overriding and super() ---")
c.show_skills()

print("\n--- Accessing Attributes ---")
print(c.father_name)
print(c.mother_name)
print(c.child_name)

print("\n--- MRO ---")
print([cls.__name__ for cls in Child.mro()])

print("\n--- isinstance() ---")
print(isinstance(c, Child))
print(isinstance(c, Father))
print(isinstance(c, Mother))
```

### Hierarchical inheritance

```python
class phone:
    def __init__(self,price,brand,camera):
        print("inside the phone constructor")
        self.price=price
        self.brand=brand
        self.camera=camera
    def buy(self):
        print("price is:",self.price)
        print("Buying a phone")
        print("we are in the phone class instance methode")


class smartphone(phone):
    pass

class featurephone(phone):
    pass

obj1=smartphone(100000,"apple","13px")
obj1.buy()
print("=====================================")
obj2=featurephone(20000,"realme","50px")
obj2.buy()
```

```python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
        print("Father constructor called")

    def show_father(self):
        print("Father Name:", self.father_name)

    def show_property(self):
        print("Father owns a house")


class Son(Father):
    def __init__(self, father_name, son_name):
        super().__init__(father_name)
        self.son_name = son_name
        print("Son constructor called")

    def show_son(self):
        print("Son Name:", self.son_name)

    def show_property(self):
        super().show_property()
        print("Son owns a bike")


class Daughter(Father):
    def __init__(self, father_name, daughter_name):
        super().__init__(father_name)
        self.daughter_name = daughter_name
        print("Daughter constructor called")

    def show_daughter(self):
        print("Daughter Name:", self.daughter_name)

    def show_property(self):
        super().show_property()
        print("Daughter owns a laptop")


# Create objects
s = Son("Anandrao", "Pratik")
d = Daughter("Anandrao", "Priya")

print("\n--- Son Details ---")
s.show_father()
s.show_son()
s.show_property()

print("\n--- Daughter Details ---")
d.show_father()
d.show_daughter()
d.show_property()

print("\n--- Attributes ---")
print(s.father_name, s.son_name)
print(d.father_name, d.daughter_name)

print("\n--- MRO ---")
print([cls.__name__ for cls in Son.mro()])
print([cls.__name__ for cls in Daughter.mro()])

print("\n--- isinstance() ---")
print(isinstance(s, Son))
print(isinstance(s, Father))
print(isinstance(d, Daughter))
print(isinstance(d, Father))
```

### Aggregation:Has-A relationship

- aggregation is a type of class relationship where one class own's another property class as a property
- the object of one class is stored inside another class as an attribute


- ex:customer has a Address,Address is a property of customer

```python
class customer:
    def __init__(self,name,gender,address):
        self.name=name
        self.gender=gender
        self.address=address

    def display_info(self):
        print("customer information:")
        print("Name is:{} ,Gender:{}".format(self.name,self.gender))
        print("The customer address is:{},{},{}".format(self.address.city,self.address.state,self.address.pin))


class Address:
    def __init__(self,city,pin,state):
        self.city=city
        self.pin=pin
        self.state=state

add1=Address("pune",411001,"Mharashtra")
cust1=customer("pratik","male",add1)
cust1.display_info()


```

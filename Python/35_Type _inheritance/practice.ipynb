Create a parent class named Person with an _init() that takes name and age, and a method introduce(). Create a child class named Doctor with its own __init() that takes name, age and specialization, and uses super().init_() to set name and age. Create a Doctor object and call introduce() along with printing the specialization.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def introduce(self):
        print("my name is {} and age {}".format(self.name,self.age))

class Doctor(Person):
    def __init__(self,name,age,specialization):
        super().__init__(name,age)
        self.specialization=specialization

d = Doctor("Pratik", 25, "Cardiologist")

d.introduce()
print("Specialization:", d.specialization)
```

Create a parent class named Ticket with an _init_() that takes passenger_name and fare, and a method print_ticket(). Create two child classes, BusTicket and TrainTicket, each with its own method (for example, seat_type()). Create one object of each and call both methods.

```python
class Ticket:
    def __init__(self, passenger_name, fare):
        self.passenger_name = passenger_name
        self.fare = fare

    def print_ticket(self):
        print("Passenger Name:", self.passenger_name)
        print("Fare: ₹", self.fare)


class BusTicket(Ticket):
    def seat_type(self):
        print("Seat Type: Sleeper")


class TrainTicket(Ticket):
    def seat_type(self):
        print("Seat Type: AC 3-Tier")


# Create objects
bus = BusTicket("Pratik", 500)
train = TrainTicket("Rahul", 1200)

# Bus ticket details
print("--- Bus Ticket ---")
bus.print_ticket()
bus.seat_type()

# Train ticket details
print("\n--- Train Ticket ---")
train.print_ticket()
train.seat_type()
```

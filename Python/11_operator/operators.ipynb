## Built in Function - 1. all( )

```python
age=25
nationality="indian"
```

```python
age>=18
```

```python
nationality=="indian"
```

```python
all((True,True))
```

```python
all((age>=18, nationality=="indian"))
```

```python
result=all((age>=18, nationality=="indian"))
print(result)
if result:
    print("you are eligible for voting")
else:
    print("you are not eligible for voting")
```

```python
a=1
b=2
c=3
```

```python
all((a<5,b<5,c<5))
```

#### 2. any()

```python
weekend=False
holiday=False
```

```python
result= any((weekend,holiday))
print(result)
if result:
    print("you can relax")
else:
    print("you need to work")
```

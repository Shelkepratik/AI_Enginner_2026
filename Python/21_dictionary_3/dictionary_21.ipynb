### Built in function on dictionary

1)len()-len() function is used to count the key-value pair in dictionary

```python
dict1={
    "name":"pratik",
    "age":23
}
print(dict1)
print(len(dict1))
```

2)min()- return minimum key in dictionary

```python
dict2={10:"a",
       20:"b",
       30:"c"
       }
print(min(dict2))
```

3)max()-return maximum key in dictionary which is numeric

```python
print(max(dict2))
```

4)sum()-it return aaddition of all keys which are numeric

```python
dict3={10:"a",
       20:"b",
       30:"c"}
print(dict3)
print(sum(dict3))
```

5)any()-any() return "True" if atleast one key is True

```python
dict4={0:"a",
       0:"b",
       0:"c",
       False:"d"}
print(dict4)
print(any(dict4))
```

```python
dict4={0:"a",
       0:"b",
       1:"c",
       False:"d"}
print(dict4)
print(any(dict4))
```

6)all()-all() function return True if all keys is True

```python
dict5={
    1:"a",
    2:"b",
    3:"c",
    4:"d"
}
print(dict5)
print(all(dict5))
```

7)sorted()-it return sorted keys in the list within the  dictionary considering key.

```python
dict6={
    40:"a",
    20:"b",
    10:'c'

}
list1=sorted(dict6)
```

```python
print(list1)
```

```python
print(dict6[10])
```

```python
for key in list1:
    print(key,":",dict6[key])
```

8)revered()-it's return keys in reversed order it provide the list

```python
dict7={
    40:"a",
    20:"b",
    10:'c'
}
list1=list(reversed(dict7))
print(list1)
```

```python
print(reversed(dict7))#it's return the object
```

```python
for key in list1:
    print(key,":",dict7[key])
```

Using the marks dictionary from Q2, print the value of the smallest key and the value of the largest key.

```python
marks = {101: "A", 105: "B", 103: "C"}

smallest = min(marks)
largest = max(marks)

print("Value of smallest key:", marks[smallest])
print("Value of largest key:", marks[largest])
```

Using the dict9 dictionary from Q8, use a for loop with sorted() to print each key and its value in ascending order of keys.

```python
list1=sorted(dict9)

for key in list1:
    print(key,":",dict9[key])
```

Create a dictionary dict10 = {10: "a", 20: "b", 30: "c"}. Print the keys in reverse order using list(reversed(dict10)), then use a for loop with reversed() to print each key and its value.

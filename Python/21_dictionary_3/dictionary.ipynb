## property

.

| Property                    | Syntax               | Example                        | Description                                                                      |
| --------------------------- | -------------------- | ------------------------------ | -------------------------------------------------------------------------------- |
| Key-value pairs             | `{key: value}`       | `d = {"name": "Pratik"}`       | Stores data as key-value pairs.                                                  |
| Ordered                     | `dict`               | `d = {"a": 1, "b": 2}`         | Maintains insertion order (Python 3.7+).                                         |
| Mutable                     | `dict[key] = value`  | `d["age"] = 22`                | Values can be added or changed after creation.                                   |
| Unique keys                 | `dict`               | `d = {"a": 1, "a": 2}`         | Keys must be unique; duplicate keys overwrite earlier values.                    |
| Duplicate values allowed    | `dict`               | `d = {"a": 10, "b": 10}`       | Different keys can have the same value.                                          |
| Keys must be hashable       | `{key: value}`       | `d = {1: "One"}`               | Keys must be hashable, such as strings, numbers, or tuples of hashable elements. |
| Values can be any data type | `{key: value}`       | `d = {"marks": [80, 90]}`      | Values can be lists, sets, dictionaries, or other objects.                       |
| Access using keys           | `dict[key]`          | `d["name"]`                    | Retrieves a value using its key.                                                 |
| Dynamic size                | `dict[key] = value`  | `d["city"] = "Pune"`           | Items can be added or deleted.                                                   |
| Nested dictionary           | `dict` inside `dict` | `d = {"student": {"age": 22}}` | A dictionary can contain another dictionary.                                     |
| Created using curly braces  | `{}`                 | `d = {"a": 1}`                 | Curly braces are used to create a dictionary.                                    |
| Empty dictionary            | `{}` or `dict()`     | `d = {}`                       | Creates a dictionary with no items.                                              |

Remember: A Python dictionary is an ordered, mutable collection of key-value pairs with unique keys.

## function



| Function   | Syntax           | Example                            | Description                                           |
| ---------- | ---------------- | ---------------------------------- | ----------------------------------------------------- |
| `len()`    | `len(dict)`      | `len({"a": 1, "b": 2})`            | Returns the number of key-value pairs.                |
| `dict()`   | `dict(iterable)` | `dict([("a", 1), ("b", 2)])`       | Creates a dictionary.                                 |
| `str()`    | `str(dict)`      | `str({"a": 1})`                    | Converts the dictionary representation into a string. |
| `type()`   | `type(dict)`     | `type({"a": 1})`                   | Returns the data type of the object.                  |
| `sorted()` | `sorted(dict)`   | `sorted({"c": 3, "a": 1, "b": 2})` | Returns a sorted list of dictionary keys.             |
| `any()`    | `any(dict)`      | `any({"a": 0, "b": 1})`            | Returns `True` if at least one key is truthy.         |
| `all()`    | `all(dict)`      | `all({"a": 1, "b": 2})`            | Returns `True` if all keys are truthy.                |
| `max()`    | `max(dict)`      | `max({"a": 1, "c": 3, "b": 2})`    | Returns the largest key.                              |
| `min()`    | `min(dict)`      | `min({"a": 1, "c": 3, "b": 2})`    | Returns the smallest key.                             |
| `sum()`    | `sum(dict)`      | `sum({1: "a", 2: "b"})`            | Returns the sum of numeric keys.                      |

Note: These are Python built-in functions that can be used with dictionaries. Dictionary methods include `keys()`, `values()`, `items()`, `get()`, and `update()`.

## method



| Method         | Syntax                          | Example                        | Description                                                              |
| -------------- | ------------------------------- | ------------------------------ | ------------------------------------------------------------------------ |
| `clear()`      | `dict.clear()`                  | `d.clear()`                    | Removes all items from the dictionary.                                   |
| `copy()`       | `dict.copy()`                   | `d2 = d.copy()`                | Creates a shallow copy of the dictionary.                                |
| `fromkeys()`   | `dict.fromkeys(keys, value)`    | `dict.fromkeys(["a", "b"], 0)` | Creates a new dictionary with given keys and a common value.             |
| `get()`        | `dict.get(key)`                 | `d.get("name")`                | Returns the value for a key, or `None` if the key is missing by default. |
| `items()`      | `dict.items()`                  | `d.items()`                    | Returns all key-value pairs.                                             |
| `keys()`       | `dict.keys()`                   | `d.keys()`                     | Returns all keys.                                                        |
| `values()`     | `dict.values()`                 | `d.values()`                   | Returns all values.                                                      |
| `pop()`        | `dict.pop(key)`                 | `d.pop("age")`                 | Removes the specified key and returns its value.                         |
| `popitem()`    | `dict.popitem()`                | `d.popitem()`                  | Removes and returns the last inserted key-value pair.                    |
| `setdefault()` | `dict.setdefault(key, default)` | `d.setdefault("city", "Pune")` | Returns the value of a key; inserts the default if the key is missing.   |
| `update()`     | `dict.update(other)`            | `d.update({"age": 22})`        | Adds new items or updates existing key values.                           |

Note: Assume `d = {"name": "Pratik", "age": 22}` for examples using `d`. Each example is independent.

#### Basic set operation

1)mutability-set is mutable 

string,tuple-immutable

list,set,dict-mutable

int,float,complex,bool-immutable

```python
set1={10,20,30}
set1.add(40)
print(set1)
```

2)Conversion -we can convert string,list,tuple,dictonary,range() in set


using Built-in function-set()


a)converting string to set

```python
str1="hello"
set1=set(str1)
print(set1)
```

b)converting list into set

```python
list1=[10,20,30]
set2=set(list1)
print(set2)
```

c)converting tuple into set

```python
tuple1=(1,2,3,4,5)
set3=set(tuple1)
print(set3)
```

d)converting dictionary into set

```python
dict1={
    "name":"pratik",
    "age":23
}
set4=set(dict1)
print(set4)
```

e)converting range() into set

```python
range1=range(1,11,1)
set5=set(range1)
print(set5)
```

3)Aliasing

```python
set1={1,2,3,4}
set2=set1
print(set1)
print(set2)
print(id(set1),id(set2))
```

4)cloning(deep copy)

```python
set1={10,20,30}
set2=set1.copy()
print(set1)
print(set2)
print(id(set1),id(set2))
```

5)identity operation("is" operator)

```python
set1={10,20,30}
set2=set1
print(set1 is set2)#same memory location in stored means same id
print(set2 is set1)
print(id(set1),id(set2))
print(id(set1)==id(set2))
```

```python
set1={10,20,30}
set2=set1.copy()
print(set1 is set2)#stored in different memeory location
print(id(set1),id(set2))
```

6)searching-"in" operator

```python
set1={10,20,30}
print(10 in set1)
print(50 in set1)
```

7)Emptiness-not operator-check wheter set is empty or not

```python
set1=set()
print(not set1)
set2={10,20,30}
print(not set2)
```

8-comparison operation-

```python
set1={1,2,3}
set2={4,5,6}
print(set1<set2)
print(set1>set2)
```

### Built in fuction on set:-

```python
#1)len()-return length of set
set1={10,20,30}
print(len(set1))
```

```python
#2)max()-return maximum element
print(max(set1))
#set2={"pratik",10,20,True}
#print(max(set2))
```

```python
#3)min()-return minimum element
print(min(set1))
```

```python
#4)sum()-total of all element
print(sum(set1))
```

```python
#5)any()-if return True if any element is true of set
set1={0,0,0,1}
print(any(set1))
```

```python
#all()-if return True all element are True in the set
set1={10,20,30,40}
print(all(set1))
set2={0,0,0,1}
print(all(set2))
```

```python
#sorted()-it return the sorted element,and it's return the list
set1={10,50,40,20,30}
print(sorted(set1))#return list
print(set(sorted(set1)))
```

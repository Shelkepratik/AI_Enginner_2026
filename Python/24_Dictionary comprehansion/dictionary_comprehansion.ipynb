### dictionary Comprehension

dictionary comprehenstion create a new dictionary by iterating over a sequence,


syntax={key expresion:value expresion for element/item in iterable if condition}

e.x-create a dictionary having squares of number from 1 to 5 where key is number and value is square

```python
dict1={i:i**2 for i in range(1,6,1)}
print(dict1)
```

e.x create a dictionary where key is the animal and value is len of string

```python
animals=["cat","dog","fish"]
dict2={i:len(i) for i in animals}
print(dict2)
```

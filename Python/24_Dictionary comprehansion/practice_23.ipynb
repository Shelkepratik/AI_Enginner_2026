Using dictionary comprehension, create a dictionary having numbers from 1 to 5 as keys and their squares as values.

```python
dict1={i:i**2 for i in range(1,6,1)}
print(dict1)
```

Create a list animals = ["cat", "dog", "fish"]. Using dictionary comprehension, create a dictionary where the key is the animal name and the value is the length of the name.

```python
animals=["cat","dog","fish"]
dict1={i:len(i) for i in animals}
print(dict1)
```

Using dictionary comprehension, create a dictionary having numbers from 1 to 10 as keys and their cubes as values, only for even numbers.

```python
dict1={i:i**3 for i in range(1,11,1) if i%2==0}
print(dict1)
```

Create a list fruits = ["apple", "banana", "mango"]. Using dictionary comprehension, create a dictionary where the key is the fruit name and the value is the fruit name in uppercase.

```python
fruits=["apple","banana","mango"]
dict1={i:i.upper() for i in fruits}
print(dict1)
```

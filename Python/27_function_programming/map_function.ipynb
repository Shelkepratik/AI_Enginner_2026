### Map Function

- map function in python is useful function that applies given function to each item of iterable(list,tuple,range) and create a new iterable with resukt

- syntax

map(function,iterable)
- function-normal fun,lambda fun 
- iterable-list,tuple,set,range

```python
def sequare(x):
    return x**2

x = [1, 2, 3, 4, 5]

result = map(sequare, x)

print(list(result))
```

- when we apply map function it return we get map object that we convert into iterable into list,set,tuple

```python
prices = [100, 200, 300, 400]

result = list(map(lambda x: x * 0.9, prices))

print(result)
print(tuple(result))
```

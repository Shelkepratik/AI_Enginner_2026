### - Lambda function

- lambda is a anonoymons function-nameless function

- it is define using lambda keyword

- lambda function is a simple throwaway function that we are not going to use in the future

#### characteristics 

- anonymous-nameless
- single expression-one line
- return value-automatically return value

syntax
- lambda arguments:expression

e.x - cretae a simple lambda function that adds two number

1)using normal function

```python
def sum(x,y):
    return x+y
result=sum(3,5)
print(result)
```

2)with lambda function

```python
add=lambda x,y:x+y
print(add(3,5))
```

e.x- check number is even or not

```python
even=lambda x:x%2==0
print(even(2))
```

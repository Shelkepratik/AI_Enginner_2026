Write a normal function summation(x, y) that returns the sum of two numbers. Then write the same logic using a lambda function and print the result for 3 and 5.

- using normal function

```python
def sum(a,b):
    return a+b
sum(3,5)
```

- with lambda function

```python
sum=lambda a,b:a+b
print(sum(3,5))
```

Write a lambda function that checks whether a number is even. Print the result for 6 and 7.

```python
even=lambda x:x % 2==0
print(even(6))
print(even(7))
```

Write a lambda function that takes three numbers and returns the largest number using max(). Print the result for 10, 45 and 23.

```python
largest=lambda x,y,z:max(x,y,z)
print(largest(10,45,23))
```

 Create a list numbers = [1, 2, 3, 4, 5]. Write a normal function square(x) and use map() to get the squares of all the numbers. First print the map object directly, then print it after converting it into a list.


```python
def sequare(x):
    return x**2

x = [1, 2, 3, 4, 5]

result = map(sequare, x)

print(list(result))
```

Using the numbers list from Q7, use map() with a normal function add_ten(x) to add 10 to every value. Then solve the same problem using map() with a lambda function.

```python
x=[1,2,3,4,5]
def add_ten(x):
    return x + 10

result = list(map(add_ten, x))

print(result)
```

```python
numbers = [10, 20, 30, 40, 50]

result = list(map(lambda x: x + 10, numbers))

print(result)
```

Create a list names = ["rahul", "amit", "sneha"]. Use map() with a lambda function to convert all the names into uppercase and print the result as a list.

```python
name=["rahul","amit","sneha"]
result=list(map(lambda x:x.upper(),name))

#map(function,iterable)

print(result)
```

Create a list prices = [100, 200, 300, 400]. Use map() with a lambda function to apply a 10% discount on every price. Print the result as a list, then as a tuple.

```python
prices = [100, 200, 300, 400]

result = list(map(lambda x: x * 0.9, prices))

print(result)
print(tuple(result))
```

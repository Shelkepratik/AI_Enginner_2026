### Reduce function

In Python, reduce() applies a function cumulatively to the items of an iterable, reducing them to one final value.

reduce(function,iterable)

```python
from functools import reduce
number=[1,2,3,4,5]
def add(x,y):
    return x+y
result=reduce(add,number)
print(result)
```

- using lambda function

```python
from functools import reduce
number=[1,2,3,4,5]
total=reduce(lambda x,y:x+y,number)
print(total)
```

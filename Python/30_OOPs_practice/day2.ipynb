6. Create a class named Product with attributes name and price. Add a method apply_discount(percent) that reduces the price by the given percent and updates self.price. Print the price before and after calling the method.

```python
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def apply_dicount(self,percent):
        discount=self.price*percent/100
        self.price=self.price-discount

p1=product("Laptop",50000)
print("Price before discount:",p1.price)
p1.apply_dicount(10)
print("price after discount:",p1.price)
```

8. Create a class named Temperature that takes celsius as a parameter. Add a method to_fahrenheit() that returns (celsius × 9/5) + 32, and another method show() that calls self.to_fahrenheit() inside it and prints both values.


```python
class Temperature:
    def __init__(self,celsius):
        self.celsius=celsius
    def to_fahrenheit(self):
        return (self.celsius*9/5)+32
    def show(self):
        print("Temperature in celsius:",self.celsius)
        print("The Temperature:",self.to_fahrenheit())
t1=Temperature(30)

t1.show()
    
```

9. Create a class named Calculator that takes two numbers as parameters. Add methods add(), subtract() and multiply(). Add one more method show_all() that calls all three methods using self and prints their results.

```python
class Calculator:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        return self.a+self.b
    def subtract(self):
        return self.a-self.b
    def multiply(self):
        return self.a*self.b
    def show_all(self):
        print("The addition is:",self.add())
        print("The substraction is:",self.subtract())
        print("The multiplication is:",self.multiply())
c1=Calculator(2,3)
c1.show_all()
```

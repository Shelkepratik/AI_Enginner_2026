create a class named Rectangle that takes length and breadth as parameters.Add two methods:area() that return length*breadth and perimeter() that returns 2*(length+breadth).create two rectangle objects and print their and permeter

```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
    def perimeter(self):
        return 2*(self.length+self.breadth)
r1=Rectangle(2,4)
r1.area()
```

```python
r2=Rectangle(2,4)
r2.perimeter()
```

Create a class named Pet with attributes name and sound. Add a method speak() that prints "<name> says <sound>". Create objects for a dog, a cat and a cow, and call speak() for each.

```python
class pet:
    def __init__(self,name,sound):
        self.name=name
        self.sound=sound
    def speak(self):
        print(f"{self.name} says {self.sound}")
dog=pet("Dog","Woof")
cat=pet("Cat","meow")
cow=pet("Cow","moo")

dog.speak()
cat.speak()
cow.speak()
```

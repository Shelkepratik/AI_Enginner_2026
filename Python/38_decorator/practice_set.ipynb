Write a decorator named add_border that prints a line of 20 "*" before and after the original function runs. Apply it to a function show_title() that prints "PYTHON NOTES". First apply it using show_title = add_border(show_title), then rewrite it using the @ syntax.

```python
def add_border(func):
    def inner():
        print("*"*20)
        func()
        print("*"*20)
    return inner

def show_title():
    print("PYTHON NOTES")

result=add_border(show_title)
result()
```

### using @ symbol

```python
def add_border(func):
    def inner():
        print("*"*20)
        func()
        print("*"*20)
    return inner

@add_border
def show_title():
    print("PYTHON NOTES")

show_title()
```

Write a decorator named log_call that prints "Function started" before calling the original function and "Function finished" after it. Apply it using @ to a function make_tea() that prints the steps to make tea.

```python

def log_call(func):
    def inner():
        print("function start")
        func()
        print("function finished")
    return inner

@log_call
def make_tea():
    print("step to make tea")

make_tea()
```

Write a function get_price() that returns 1000. Write a decorator apply_discount that returns the price after a 10% discount. Apply it using @ and print the final price.

```python
def apply_discount(func):
    def inner():
        price=func()
        return price-(price*10/100)
    return inner

@apply_discount
def get_price():
    return 1000

get_price()

```

Write a decorator to_title that converts the returned string into title case. Apply it to two different functions: get_course() returning "python for data science" and get_city() returning "new delhi". Print both results.

```python
def to_title(func):
    def inner():
        result=func()
        return result.title()
    return inner

@to_title
def get_course():
    return "python for data science"

@to_title
def get_city():
    return "new delhi"

print(get_course())
print(get_city())
```

Write a decorator remove_spaces that removes extra spaces from the start and end of the returned string using strip(). Apply it to a function get_username() that returns "   priya_sharma   ". Print the result along with its len() before and after decoration.

```python
def remove_spaces(func):
    def inner():
        result=func()
        return result.strip()
    return inner

@remove_spaces        
def get_username():
    return "   priya_sharma   "

get_username()
```

Write two decorators: make_upper (converts the returned string to uppercase) and add_exclaim (adds "!!!" at the end of the returned string). Apply both to a function get_message() that returns "well done". Try both orders (@make_upper above @add_exclaim and the opposite), and write a comment on whether the output changes.

```python
def make_upper(func):
    def inner():
        return func().upper()
    return inner

def add_exclaim(func):
    def inner():
        return func()+"!!!"
    return inner

@make_upper
@add_exclaim
def get_message():
    return "well done"

get_message()
```

#### using @decor

```python
def decor(func):#decor(addition)
    def inner():
        result1=func()
        num3=int(input("enter the thired number"))
        final_result=result1+num3
        return final_result
    return inner
    
@decor
def addition():
    num1=int(input("Enter first no: "))
    num2=int(input("Enter the second no"))
    result=num1+num2
    return result

print("The addition is:",addition())

```

```python
def deco1(func):
    def inner():
        return func().upper()
    return inner

def deco2(func):
    def inner():
        return func().split()
    return inner

@deco2
@deco1
def get_name():
    name=input("enter the name: ")
    last_name=input("Enter the last name")
    full_name=name+" "+last_name
    return full_name

print(get_name())
```

```python
def my_decorator(func):
    def inner():
        result=func()
        return result.upper()
    return inner


@my_decorator
def get_name():
    return 'rahul'

@my_decorator
def get_city():
    return "pune"

print(get_name())
print(get_city())
```

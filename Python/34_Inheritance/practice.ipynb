Create a parent class named Animal with methods eat() and sleep(). Create a child class named Dog that inherits from Animal and has its own method bark(). Create a Dog object and call all three methods.

```python
class Animal:
    def eat(self):
        print("eating")

    def sleep(self):
        print('sleeping')

class Dog(Animal):
    def bark(self):
        print("dog is bark")

d=Dog()
d.eat()
d.sleep()
d.bark()
```

Create a class named Employee with a method work(). Create a class named Developer that does NOT use inheritance, and write the same work() method in it again along with a method write_code(). Then rewrite Developer using inheritance. Write a comment about how much repeated code was removed.

1. Without inheritance (repeated code)

```python
class Employee:
    def work(self):
        print('Employee is working')

class Developer:
    def work(self):
        print("Employee is working")

    def write_code(self):
        print("Developer is writing code")

d=Developer()
d.work()
d.write_code()
```

2. With inheritance (code reuse)

```python
class Employee:
    def work(self):
        print("Employee is working")


class Developer(Employee):
    def write_code(self):
        print("Developer is writing code")


d = Developer()
d.work()
d.write_code()

# Repeated code removed: 2 lines (the duplicate work() method and its print statement).
```

Create a parent class named Account with an _init() that sets account_holder, and a method show_holder(). Create a child class SavingsAccount with its own __init_() that sets interest_rate only. Create a SavingsAccount object and call show_holder(). Write a comment explaining the error you get and why.

```python
class Account:
    def __init__(self, account_holder):
        self.account_holder = account_holder

    def show_holder(self):
        print("Account holder:", self.account_holder)


class SavingsAccount(Account):
    def __init__(self, interest_rate):
        self.interest_rate = interest_rate


s = SavingsAccount(5)
s.show_holder()

# Error: AttributeError because the child's __init__() does not call
# the parent's __init__(), so account_holder is never initialized.
```

Fix the program from Q8 without using super(): inside the SavingsAccount _init_(), take both account_holder and interest_rate as parameters and set both attributes using self. Create the object again and check that show_holder() now works.

```python
class Account:
    def __init__(self, account_holder):
        self.account_holder = account_holder

    def show_holder(self):
        print("Account holder:", self.account_holder)


class SavingsAccount(Account):
    def __init__(self, account_holder, interest_rate):
        self.account_holder = account_holder
        self.interest_rate = interest_rate


s = SavingsAccount("Pratik", 5)
s.show_holder()

# The error is fixed because account_holder is initialized directly
# inside the child class without using super().
```

Create a parent class named Restaurant with an _init_() that takes restaurant_name, and methods show_menu() and take_order(). Create a child class named CloudKitchen that inherits from Restaurant and adds a method home_delivery(). Create a CloudKitchen object and call all three methods to show that it can do everything a restaurant can do, plus delivery.

```python
class Restaurant:
    def __init__(self, restaurant_name):
        self.restaurant_name = restaurant_name

    def show_menu(self):
        print(self.restaurant_name, "- Showing menu")

    def take_order(self):
        print(self.restaurant_name, "- Taking order")


class CloudKitchen(Restaurant):
    def home_delivery(self):
        print(self.restaurant_name, "- Providing home delivery")


# Create a CloudKitchen object
c = CloudKitchen("Food Hub")

# Call all three methods
c.show_menu()
c.take_order()
c.home_delivery()
```

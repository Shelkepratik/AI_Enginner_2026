### Inheritance
- IS-A relationship
- Inheritance allows a child class to acquire the attribute and method of a parent class-just like a child inherits property from parents in real life

#### parent class:
- Base class
- super class
- The one that gives

#### child class
- derived class
- sub class
- that one recievs


- child has no constructore

- super() keyword:we access the parent class constructor in child class that time we use the super() method

```python
class Device:
    def __init__(self,price,brand):
        self.price=price
        self.brand=brand


class Smartphone(Device):
    def __init__(self,price,brand,os,ram):
        super().__init__(price,brand)
        self.os=os
        self.ram=ram



phone1=Smartphone(18000,"realme13","ios",8)
print(phone1.brand)
print(phone1.price)
print(phone1.os)
print(phone1.ram)
```

```python
class AppUser:
    def __init__(self,username):
        self.username=username

    def login(self):
        print(self.username,"logged in!")

class student(AppUser):
    def __init__(self,roll_no,username):
        super().__init__(username)
        self.roll_no=roll_no

    def enroll(self):
        print("enroll successfully")

student1=student(100,"pratik")
student1.enroll()
student1.login()
        
```

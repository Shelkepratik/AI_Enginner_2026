Write a program that reads shopping.txt and prints only the first line using readline(), and then only the first 10 characters using read(10).

```python
with open("shopping1.txt","r") as file:
    result=file.readline()
    print("first line: ",result)

#Read only the first 10 character
with open("shopping1.txt","r") as file:
    result=file.read(10)
    print("First 10 character:",result)
```

Create a file named marks.txt with the content "Maths 85", "Science 90" and "English 78" on separate lines. Write a program that reads the file and prints the total number of lines in it.

```python

# Create and write to the file
with open("marks.txt", "w") as file:
    file.write("Maths 85\n")
    file.write("Science 90\n")
    file.write("English 78\n")

# Read the file and count the lines
with open("marks.txt", "r") as file:
    lines = file.readlines()
    print("Total number of lines:", len(lines))
```

Write a program that reads any text file and counts the total number of words in it using split().

```python
with open("marks.txt","r") as file:
    content=file.read()
    words=content.split()
    print("Total number of words: ",len(words))
```

Write a program that asks the user for a file name and prints its content. Use try and except to handle FileNotFoundError if the file does not exist.

```python
file_name=input("enter file name")
try:
    with open(file_name,"r") as file:
        content=file.read()
        print(content)
except FileNotFoundError:
    print("error:file not found")
```

Before creating a file named report.txt, use os.path.exists() to check whether it already exists. If it exists, print "File already exists", otherwise create it and write "New report created" in it.

```python

import os

file_name = "report.txt"

if os.path.exists(file_name):
    print("File already exists")
else:
    with open(file_name, "w") as file:
        file.write("New report created")
    print("File created successfully")
```

Using the os module, create a folder named python_notes, print the current working directory, and then print the list of all files and folders in the current working directory.

```python

import os

# Create a folder
os.makedirs("python_notes", exist_ok=True)

# Print the current working directory
print("Current working directory:", os.getcwd())

# Print all files and folders
print("Files and folders:", os.listdir())
```

Create a folder structure projects/2026/october using os.makedirs(). Create a file named tasks.txt inside the october folder (build its path using os.path.join()). Then make a copy of it named tasks_backup.txt using shutil.copy(), rename the original to tasks_final.txt using os.rename(), and finally delete the backup file using os.remove().

```python

import os
import shutil

# 1. Create the folder structure
os.makedirs("projects/2026/october", exist_ok=True)

# 2. Build the file path using os.path.join()
file_path = os.path.join("projects", "2026", "october", "tasks.txt")

# 3. Create the file and write content
with open(file_path, "w") as file:
    file.write("Complete Python practice")

print("tasks.txt created")

# 4. Make a copy of the file
backup_path = os.path.join("projects", "2026", "october", "tasks_backup.txt")
shutil.copy(file_path, backup_path)

print("Backup created")

# 5. Rename the original file
final_path = os.path.join("projects", "2026", "october", "tasks_final.txt")
os.rename(file_path, final_path)

print("Original file renamed")

# 6. Delete the backup file
os.remove(backup_path)

print("Backup deleted")
```

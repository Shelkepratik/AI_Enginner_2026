4.Aliasing(shallow copy)-In Aliasing ,multiple variable refered to the same list object in memory .

```python
list1=[10,20,30]
print(list1)
print(id(list1))

list2=list1 #Aliasing apply
print(list2)
print(id(list2))

list2[0]=5
print(list1)
print(list2)
```

5.cloning(deep copy)-in cloning ,we copy content of one list into another list
-after cloning ,both lists will be diffent even if both content is some
-if we change one list ,it will not reflect that change into another list,it is also called as deep copy

```python
list1=[10,20,30]
print(list1)
print(id(list1))
list2=list1.copy()
print(list2)
print(id(list2))
list2[0]=5
print(list2)
print(list1)
```

6.searching-"in" operator used."in"operator is used to check wheter elemenet is present in the list ,element is present in the list it's return "True" otherwise "false"

```python
list1=[10,10.1,2+3j,"ganesh",True]
print(len(list1))

print(10 in list1)

print(2+3j in list1)

print("om" in list1)

print("om" not in list1)

print(True not in list1)
```

7.identity-"is" operator.check wheter two variable is refer to same memory location

```python
list1=[10,20,30]
list2=list1
print(id(list1),id(list2))
print(list1 is list2)
print(list2 is list1)


list3=list1.copy()
print(id(list3),id(list1))
print(list3 is list1)
print(list1 is list3)
print(list2 is list3)
```

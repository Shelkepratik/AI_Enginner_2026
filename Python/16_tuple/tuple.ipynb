#### 6-Accessing tuple element

we can print and access entire tuple using tuple name

```python
tuple7=(10,20,30,40,50)
print(tuple7)
```

To access the individual element of the tuple we use indexing

-these are two type of indexing

1-positiveindexing-starts from 0,left to right

2-negative indexing-start from -1,right to left

```python
tuple8=(10,20,30,40,50)
print(tuple8[0])
print(tuple8[1])
print(tuple8[2])
print(tuple8[3])
print(tuple8[4])
```

```python
print(tuple8[-1])
print(tuple8[-2])
print(tuple8[-3])
print(tuple8[-4])
print(tuple8[-5])
```

we can access the subtuple or group of element of tuple using slicing.

```python
#syntax=tuple_name[start:end+1:step]
tuple10=[10,20,30,40,50]
print(tuple10[0:1])
print(tuple10[:1:1])
print(tuple10[1:4:1])
print(tuple10[::])
print(tuple10[::2])
print(tuple10[0:5:1])
print(tuple10[0:5:2])
```

```python
print(tuple10)
print(tuple10[-4:-1:1])
print(tuple10[4:1:-1])
print(tuple10[4:2:-1])
```

#### 7-looping in tuple

we want to process each element of two tuple,we have to iterate tuple ,and we perform this using while loop and for loop

1-while loop

```python
tuple=(10,20,30)
i=0
while i<len(tuple):
    print(tuple[i])
    i+=1
```

2-for loop

```python
tuple=(10,20,30)
for i in tuple:
    print(i,end=" ")
```

3-enumerate() function-if we want to print index number as well we use enumerate function

```python
tuple=(10,20,30)
for index,var in enumerate(tuple):
    print(index,var)
```

### Basic tuple operation 

1-Immutability-tuple is immutable,unchange.

-once we defined/create a tuple we caanot add,remove,replace/modify element of the tuple.

```python
tuple1=(10,20,30)
tuple1[0]=11
print(tuple1)
```

2-concatenation-we can join one tuple at the end of another tuple using  "+" operator

```python
tupl1=(10,20,30)
tuple2=(40,50,60)
tuple=tuple1+tuple2
print(tuple)
```

3-conversion-we can convert string,list,set,dictinary,range() into tuple,so we use tuple() function 

a-converting string to tuple

```python
del(tuple)
```

```python
str1="Hello"
tuple1=tuple(str1)
print(tuple1)
```

b-converting list to tuple

```python
list1=[10,20,30]
tuple2=tuple(list1)
print(tuple2)
```

c-converting set to tuple

```python
set1={10,20,30}
tuple3=tuple(set1)
print(tuple3)
```

d=converting dictionary to tuple

```python
dict1={"name":"ganesh",
     "age":23}
tuple4=tuple(dict1)
print(tuple4)
```

e-converting range() to tuple

```python
range1=range(1,11,1)
tuple5=tuple(range1)
print(tuple5)
```

4-Aliasing-aliasing mans multiple variable referring same tuple object in memory

```python
tuple1=(10,20,30)
tuple2=tuple1
print(tuple2)
print(tuple1)
print(id(tuple1),id(tuple2))
```

5-cloning(not allow):-using cloning we copy content of one tuple into another tuple,after copying we have same element in both the tuple but both refer to the different memory location

```python
tuple1=(10,20,30)
tuple2=tuple1.copy()#not allow
print(tuple1)
print(tuple2)
```

6-searching ("in" operator):using in operator we can check if particular element is present in the tuple or not

```python
tuple1=(10,10.1,True,2+3j,"pratik")
print(10 in  tuple1)
print(20 in tuple1)
```

6-identity("is"operator)-using identity operator we check if two variable refer to the same memory location

```python
tuple1=(10,20,30)
tuple2=tuple1
print( tuple1 is tuple2)
```

7-comaparision-we can compare content of tuple we compare it element wise

```python
tuple1=(10,20,30,40)
tuple2=(10,20,50,60)
print(tuple1<tuple2)
print(tuple1>tuple2)
```

8-Emptiness-using not operator ,we can check if tuple is empty or not

```python
tuple1=()
print(not tuple1)
```

### Built in function in tuple

1.len()-it returns count of element in the tuple

```python
tuple1=(10,20,30)
print(len(tuple1))
```

2-max()-it return largest element in tuple

```python
print(max(tuple1))
```

3.min()-it return smallest element in tuple.

```python
print(min(tuple1))
```

4-sum()-addition of all element

```python
print(sum(tuple1))
```

5.any()-it return true if any one element is true in the tuple.

```python
tuple2=(0,0,0,1)
print(any(tuple2))
tuple3=(0,0,0,0)
print(any(tuple3))
```

6-all()-its return true if all the element of the tuple is true if any one element is false it will be return false

```python
tuple4=(10,20,30,40)
print(all(tuple4))
tuple5=(10,20,30,0)
print(all(tuple5))
```

7.sorted()-it return tuple in sorted(asscending) order.


```python
tuple6=(50,30,20,40,10)
print(sorted(tuple6))#it return list
print(tuple(sorted(tuple6)))
```

8reversed()-it return tiple in reversed order

```python
tuple7=(10,20,30,40,50)
result=reversed(tuple7)
print(result)
print(tuple(result))
```

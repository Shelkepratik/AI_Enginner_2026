## function



| Function      | Syntax             | Example                       | Description                                       |
| ------------- | ------------------ | ----------------------------- | ------------------------------------------------- |
| `len()`       | `len(tuple)`       | `len((10, 20, 30))`           | Returns the number of elements.                   |
| `max()`       | `max(tuple)`       | `max((10, 20, 30))`           | Returns the largest element.                      |
| `min()`       | `min(tuple)`       | `min((10, 20, 30))`           | Returns the smallest element.                     |
| `sum()`       | `sum(tuple)`       | `sum((10, 20, 30))`           | Returns the sum of numeric elements.              |
| `sorted()`    | `sorted(tuple)`    | `sorted((30, 10, 20))`        | Returns a new sorted list.                        |
| `tuple()`     | `tuple(iterable)`  | `tuple([1, 2, 3])`            | Converts an iterable into a tuple.                |
| `any()`       | `any(tuple)`       | `any((0, 0, 5))`              | Returns `True` if at least one element is truthy. |
| `all()`       | `all(tuple)`       | `all((1, 2, 3))`              | Returns `True` if all elements are truthy.        |
| `enumerate()` | `enumerate(tuple)` | `list(enumerate(("A", "B")))` | Returns index-value pairs.                        |
| `reversed()`  | `reversed(tuple)`  | `tuple(reversed((1, 2, 3)))`  | Reverses the order through a reverse iterator.    |

Note: These are Python built-in functions that work with tuples. The tuple has only two built-in methods: `count()` and `index()`.

## method



| Method    | Syntax           | Example                      | Description                                                |
| --------- | ---------------- | ---------------------------- | ---------------------------------------------------------- |
| `count()` | `tuple.count(x)` | `(10, 20, 10, 30).count(10)` | Returns the number of times an element appears in a tuple. |
| `index()` | `tuple.index(x)` | `(10, 20, 30).index(20)`     | Returns the index of the first occurrence of an element.   |

Remember: A tuple has only 2 built-in methods because tuples are immutable (their elements cannot be changed after creation).

## property



| Property                    | Syntax                 | Example                  | Description                                           |
| --------------------------- | ---------------------- | ------------------------ | ----------------------------------------------------- |
| Ordered                     | `tuple`                | `t = (10, 20, 30)`       | Elements maintain their order.                        |
| Immutable                   | `tuple`                | `t = (10, 20, 30)`       | Elements cannot be changed after creation.            |
| Allows duplicates           | `tuple`                | `t = (10, 10, 20)`       | A tuple can contain duplicate elements.               |
| Allows different data types | `tuple`                | `t = (10, "Hello", 2.5)` | Stores different types of values.                     |
| Indexed                     | `tuple[index]`         | `t[0]`                   | Elements can be accessed using indexes.               |
| Supports negative indexing  | `tuple[-index]`        | `t[-1]`                  | Accesses elements from the end.                       |
| Allows nested tuples        | `tuple` inside `tuple` | `t = ((1, 2), (3, 4))`   | A tuple can contain other tuples.                     |
| Created using parentheses   | `( )`                  | `t = (1, 2, 3)`          | Parentheses are commonly used to create tuples.       |
| Supports packing            | `tuple` assignment     | `t = 10, 20, 30`         | Multiple values can be packed into a tuple.           |
| Supports unpacking          | `a, b = tuple`         | `a, b = (10, 20)`        | Tuple elements can be assigned to separate variables. |
| Empty tuple supported       | `()`                   | `t = ()`                 | Creates a tuple with no elements.                     |
| Single-element tuple        | `(value,)`             | `t = (10,)`              | A comma is required to create a single-element tuple. |

Remember: The main properties of a tuple are ordered, immutable, and allow duplicates.

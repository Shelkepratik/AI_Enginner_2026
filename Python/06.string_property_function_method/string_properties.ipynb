#### 2. Concatenation (+)

```python
greeting="Hello"
name="Ganesh"
space=" "
```

```python
print(id(greeting))
print(id(name))
```

```python
#"HelloGanesh"
message=greeting+space+name
print(message)
```

```python
print(id(message))
```

```python
greeting="Hello "
name="Ganesh"
print(greeting+name)
```

```python
print("Hello "+"Ganesh")
```

#### 3. repetition (*)

```python
word="hi "
```

```python
repeated_word=word*3
```

```python
print(repeated_word)
```

#### 4. membership testing (in and not in)

```python
text="Python Programming"
```

```python
print("Python" in text)
```

```python
print("Java" in text)
```

```python
print("python" in text)
```

```python
print("Java" not in text)
```

```python
print("Python" not in text)
```

## API's

### What is API?
- API-application programming interface.
- website=frontend+backend
- API's are mechanisum that enable two software components sunch as the frontend and backend of an application to communicate with each other,using defined set of rule,protocol and data formate

### without API-(monolithic architeatures)
- monolithic architeatures is called frontend and backend is present in same folder along with database
- Monolithic Architecture is typically coupled (tightly coupled).
- usecase-IRCTC company:
     loc1,loc2,date=>submit
     - output=list of train
- without api to create different backend for mobile,desktop,website application
- To develop the website then hire different developer i.e frontend developer,backend developer

### with API
- backend share with different website
- To connect the same backend to mobile,desktop,website application
- we can public our function all over world
- eg:www.sppu.com/student_login,here student_login is work like function
